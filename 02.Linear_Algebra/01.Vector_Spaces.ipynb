# **BAB 1: RUANG VEKTOR DAN OPERASI DASAR**

## 1. Definisi dan Penyajikan Vektor
* **Vektor** adalah objek geometri yang memiliki **besaran (magnitude/panjang)** dan **arah**.
* **Skalar** adalah objek yang hanya memiliki besaran saja tanpa arah.
* Vektor pada ruang $\mathbb{R}^n$ secara umum dapat disajikan dalam bentuk vektor kolom maupun vektor baris:
  $$\mathbf{u} = (u_1, u_2, \dots, u_n)$$
* Dalam ruang $\mathbb{R}^2$ dan $\mathbb{R}^3$, vektor dapat dinyatakan menggunakan **vektor basis standar** $\mathbf{i}, \mathbf{j}, \mathbf{k}$:
  $$\mathbf{u} = x\mathbf{i} + y\mathbf{j} + z\mathbf{k}$$

```python
import numpy as np

u_2d = np.array([4, -1])
v_2d = np.array([0, 5])
u_3d = np.array([3, 4, -1])
v_3d = np.array([-1, 2, 2])

print(f"Vektor u (2D) : {u_2d}")
print(f"Vektor v (3D) : {v_3d}\n")
```

## 2. Operasi Dasar Vektor
* **Penjumlahan dan Pengurangan**:
  $$\mathbf{u} \pm \mathbf{v} = (u_1 \pm v_1, u_2 \pm v_2, \dots, u_n \pm v_n)$$
* **Perkalian Skalar**:
  $$k\mathbf{u} = (ku_1, ku_2, \dots, ku_n)$$

```python
vektor_jumlah = u_3d + v_3d
vektor_kurang = v_2d - 3 * u_2d
k = 2
vektor_skalar = k * u_3d

print(f"u + v (3D)    : {vektor_jumlah}")
print(f"v - 3u (2D)   : {vektor_kurang}")
print(f"2 * u (3D)    : {vektor_skalar}\n")
```

## 3. Panjang Vektor, Jarak, dan Vektor Satuan
* **Panjang Vektor (Norm/Magnitude)** dari $\mathbf{v} \in \mathbb{R}^n$:
  $$\|\mathbf{v}\| = \sqrt{v_1^2 + v_2^2 + \dots + v_n^2}$$
* **Jarak Antara Dua Vektor** $\mathbf{u}$ dan $\mathbf{v}$:
  $$d(\mathbf{u}, \mathbf{v}) = \|\mathbf{u} - \mathbf{v}\| = \sqrt{(u_1 - v_1)^2 + (u_2 - v_2)^2 + \dots + (u_n - v_n)^2}$$
* **Vektor Satuan (Unit Vector)** $\mathbf{w}$ yang searah dengan $\mathbf{v}$:
  $$\mathbf{w} = \frac{\mathbf{v}}{\|\mathbf{v}\|}$$

```python
norm_u_2d = np.linalg.norm(u_2d)
norm_u_3d = np.linalg.norm(u_3d)

# Jarak antara u_3d dan v_3d
jarak_u_v = np.linalg.norm(u_3d - v_3d)

# Vektor Satuan dari u_3d
unit_u_3d = u_3d / norm_u_3d

print(f"Panjang vektor u (2D)   : {norm_u_2d:.3f}")
print(f"Panjang vektor u (3D)   : {norm_u_3d:.3f}")
print(f"Jarak u dan v (3D)      : {jarak_u_v:.3f}")
print(f"Vektor unit u (3D)      : {unit_u_3d}\n")
```

## 4. Dot Product (Perkalian Titik) dan Sudut
* **Dot Product** menghasilkan nilai skalar:
  $$\mathbf{u} \cdot \mathbf{v} = u_1 v_1 + u_2 v_2 + \dots + u_n v_n = \|\mathbf{u}\| \|\mathbf{v}\| \cos(\theta)$$
* Dua vektor dikatakan **saling tegak lurus (ortogonal)** jika $\mathbf{u} \cdot \mathbf{v} = 0$.
* **Sudut antara dua vektor**:
  $$\cos(\theta) = \frac{\mathbf{u} \cdot \mathbf{v}}{\|\mathbf{u}\| \|\mathbf{v}\|}$$

```python
dot_product = np.dot(u_3d, v_3d)
cos_theta = dot_product / (np.linalg.norm(u_3d) * np.linalg.norm(v_3d))
sudut_rad = np.arccos(cos_theta)
sudut_deg = np.degrees(sudut_rad)

print(f"Dot Product (u . v)    : {dot_product}")
print(f"Cos(Theta)             : {cos_theta:.3f}")
print(f"Sudut Theta (Derajat)  : {sudut_deg:.2f}°\n")
```

## 5. Cross Product (Perkalian Silang)
* Cross Product **hanya berlaku pada ruang $\mathbb{R}^3$** dan menghasilkan sebuah vektor yang tegak lurus terhadap kedua vektor penyusunnya.
* Jika $\mathbf{u} = (u_1, u_2, u_3)$ dan $\mathbf{v} = (v_1, v_2, v_3)$:
  $$\mathbf{u} \times \mathbf{v} = (u_2 v_3 - u_3 v_2, u_3 v_1 - u_1 v_3, u_1 v_2 - u_2 v_1)$$

```python
cross_product = np.cross(u_3d, v_3d)

print(f"Cross Product (u x v)  : {cross_product}\n")
```

## 6. Kombinasi Linier dan Kebebasan Linier
* Sebuah vektor $\mathbf{w}$ dikatakan **kombinasi linier** dari vektor $\mathbf{u}$ dan $\mathbf{v}$ jika terdapat skalar $k_1$ dan $k_2$ sedemikian hingga:
  $$\mathbf{w} = k_1\mathbf{u} + k_2\mathbf{v}$$
* Himpunan vektor $S = \{\mathbf{v}_1, \mathbf{v}_2, \dots, \mathbf{v}_n\}$ dikatakan **bebas linier** jika persamaan:
  $$k_1\mathbf{v}_1 + k_2\mathbf{v}_2 + \dots + k_n\mathbf{v}_n = \mathbf{0}$$
  hanya dipenuhi oleh koefisien **satu-satunya** (solusi trivial) yaitu $k_1 = k_2 = \dots = k_n = 0$.
* Jika ada solusi lain di mana sekurang-kurangnya satu $k_i \neq 0$, maka himpunan vektor tersebut dikatakan **bergantung linier**.

```python
A = np.array([[1, 6],[2, 4],[-1, 2]])
w = np.array([9, 2, 7])

k_sol, residuals, rank, s = np.linalg.lstsq(A, w, rcond=None) # untuk menyelesaikan sistem persamaan linier k1, k2

u1 = np.array([-3, 0, 4])
u2 = np.array([5, -1, 2])
u3 = np.array([1, 1, 8])

M = np.column_stack((u1, u2, u3))
det_M = np.linalg.det(M)

print(f"Nilai skalar kombinasi linier [k1, k2] : {k_sol}")
print(f"Determinan matriks M                   : {det_M:.2f}")
if np.abs(det_M) > 1e-9:
    print("Status Kebebasan Linier                : Bebas Linier (Determinan != 0)\n")
else:
    print("Status Kebebasan Linier                : Bergantung Linier (Determinan = 0)\n")
```

### 1. Diketahui titik awal sebuah vektor berada pada P_1 (3,5)dan titik ujungnya berada pada P_2 (2,8). Tentukan komponen vektor tersebut dan hitung panjang vektornya! 

```python
import math
import numpy as np

titik_1 = [3,5]
titik_2 = [2,8]
komponen_x = titik_2[0] - titik_1[0]
komponen_y = titik_2[1] - titik_1[1]

vector_xy = (komponen_x ** 2) + (komponen_y ** 2)
panjang_vector = math.sqrt(vector_xy)

print(f'Panjang Vector Adalah {panjang_vector:.4f}')
```

### 2. Diberikan dua buah vektor pada ruang dimensi tiga, yaitu u=(−3,1,2)dan v=(4,0,−8). Tentukan hasil dari operasi vektor 6u+2v!> 

```python
vct_u = np.array([-3,1,2])
vct_v = np.array([4,0,-8])

result_vct = (6 * vct_u) + (2 * vct_v)
print(f'Hasil Operasinya adalah {result_vct}')
```

### 3. Diberikan vektor v=(0,5,1). Tentukan vektor satuan (unit vector) yang memiliki arah sama dengan vektor v! 

```python
vct_v2 = np.array([0,5,1])
jrk_v2 = np.linalg.norm(vct_v2)
satuan_v2 = vct_v2 / jrk_v2
print(f'vector satuan dari v = (0,5,1) adalah : {satuan_v2}')
```

### 4. Hitunglah jarak antara dua titik/vektor u=(1,3,−2,7)dan v=(0,7,2,2)  dalam ruang dimensi empat (R4)!

```python
vct_u3 = np.array([1,3,-2,7])
vct_v3 = np.array([0,7,2,2])
count = 0

for i in range(4):
    count += (vct_u3[i] - vct_v3[i]) ** 2
jarak_vector = np.sqrt(count)
print(f'Jarak Vecotrnya adalah {jarak_vector:.3f}')
```

### 5. Diketahui dua buah vektor u=3i+4j−kdan v=−i+2j+2k. Hitunglah nilai hasil dari perkalian titik (u⋅v)!

```python
vct_u4 = np.array([3,4,-1])
vct_v4 = np.array([-1,2,2])
dot_mult = 0

for i in range(3):
    dot_mult += vct_u4[i] * vct_v4[i]

print(f'nilai hasil dari perkalian titik (u⋅v) adalah {dot_mult}')
```

### 6. Tentukan besar sudut θ(dalam derajat) yang dibentuk oleh dua vektor u=(0,0,1)dan v=(0,2,2)!

```python
vct_u5 = np.array([0,0,1])
vct_v5 = np.array([0,2,2])
pjg_vct_u5 = np.linalg.norm(vct_u5)
pjg_vct_v5 = np.linalg.norm(vct_v5)

sudut_0 = np.arccos((np.dot(vct_u5,vct_v5)) / (pjg_vct_u5 * pjg_vct_v5))
print(f'Besar Sudut dari kedua Vector adalah {sudut_0:.4f} derajat')
```

### 8. Selidiki apakah vektor w=(−4,8)merupakan kombinasi linier dari vektor u=(−1,3)dan v=(2,1)! Jika ya, tentukan nilai skalar k_1 dan k_2 yang memenuhi w=k_1 u+k_2 v

```python
vct_a = np.array([[-1,3],[2,1]])
vct_b = np.array([-4.,8.])

# k1 = -(vct_b[0] - (vct_a[1][0] * vct_b[1])) / -(vct_a[0][0] - (vct_a[1][0] * vct_a[0][1]))
# k2 = vct_b[1] - (vct_a[0][1] * k1)

A = vct_a.T
k1, k2 = np.linalg.solve(A, vct_b)
cek_result = (k1 * vct_a[0]) + (k2 * vct_a[1])

if np.array_equal(cek_result, vct_b):
    print(f'Nilai Skalar K1={k1:.4f} dan K2={k2:.4f} adalah hasil dari {vct_b}')
else:
    print(f'Nilai Skalar K1={k1:.4f} dan K2={k2:.4f} bukan hasil dari {vct_b}')
```

### 9. Diketahui tiga buah vektor pada ruang dimensi tiga, yaitu u=(3,1,2), v=(1,2,1), dan w=(2,−1,1). Tentukan apakah ketiga vektor tersebut bersifat bebas linier atau bergantung linier! 

```python
vct_a = np.array([3,1,2])
vct_b = np.array([1,2,1])
vct_w = np.array([2,-1,1])

M = np.column_stack((vct_a,vct_b,vct_w))
det_M = np.linalg.det(M)
if np.abs(det_M) > 1e-9:
    print("Status Kebebasan Linier : Bebas Linier (Determinan != 0)\n")
else:
    print("Status Kebebasan Linier : Bergantung Linier (Determinan = 0)\n")
```
