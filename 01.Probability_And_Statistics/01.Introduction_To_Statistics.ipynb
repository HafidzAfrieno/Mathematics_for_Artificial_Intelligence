# **BAB 1 : PENGANTAR STATISTIKA**

## 1. Definisi dan Konsep Dasar

Statistika adalah sekumpulan metode untuk mengumpulkan, menampilkan, menganalisis, dan menarik kesimpulan dari data. Pemahaman mengenai statistika terbagi menjadi dua cabang utama:
*   **Statistika Deskriptif:** Cabang statistika yang berkaitan dengan pengorganisasian, penampilan, dan pendeskripsian data.
*   **Statistika Inferensial:** Cabang statistika yang berkaitan dengan penarikan kesimpulan tentang populasi berdasarkan informasi yang terkandung dalam sebuah sampel dari populasi tersebut.

Untuk melakukan inferensi, terdapat beberapa definisi dasar yang menghubungkan teori statistika dengan dunia nyata:
*   **Populasi:** Setiap koleksi spesifik dari objek yang menjadi pusat perhatian atau ketertarikan.
*   **Sampel:** Setiap subhimpunan (subset) dari populasi. Jika sampel mencakup seluruh populasi, proses ini disebut sensus.
*   **Pengukuran dan Data Sampel:** Pengukuran adalah angka atau atribut yang dihitung untuk setiap anggota populasi atau sampel. Pengukuran elemen-elemen sampel secara kolektif disebut sebagai data sampel.
*   **Parameter:** Sebuah angka yang merangkum suatu aspek dari populasi secara keseluruhan (contoh: rata-rata populasi disimbolkan dengan $\mu$ dan proporsi populasi disimbolkan dengan $p$).
*   **Statistik:** Sebuah angka yang dihitung dari data sampel (contoh: rata-rata sampel disimbolkan dengan $\overline{x}$ dan proporsi sampel disimbolkan dengan $\hat{p}$).

Data sendiri diklasifikasikan ke dalam dua jenis pengukuran:
*   **Data Kualitatif:** Pengukuran yang tidak memiliki skala numerik alami, melainkan terdiri dari atribut, label, atau karakteristik non-numerik lainnya.
*   **Data Kuantitatif:** Pengukuran numerik yang muncul dari skala numerik alami.

## 2. Contoh Soal & Pembahasan

**Soal:**
Seorang peneliti ingin memperkirakan rata-rata jumlah uang yang dihabiskan per orang oleh pengunjung sebuah taman hiburan. Ia mengambil sampel acak sebanyak 40 pengunjung dan mendapatkan rata-rata sebesar $\$28$ per orang. Tentukan:
1. Apa populasi yang menjadi perhatian?
2. Apa parameter yang menjadi perhatian?
3. Apa statistik yang terlibat, dan berdasarkan sampel ini, apakah kita mengetahui rata-rata pasti dari populasi?

**Pembahasan:**
*   **Langkah 1 (Menentukan Populasi):** Populasi adalah keseluruhan objek penelitian. Dalam kasus ini, populasinya adalah seluruh pengunjung taman hiburan tersebut.
*   **Langkah 2 (Menentukan Parameter):** Parameter adalah ukuran atau angka yang mendeskripsikan keseluruhan populasi. Parameternya adalah rata-rata uang yang dihabiskan oleh *seluruh* pengunjung taman hiburan.
*   **Langkah 3 (Menentukan Statistik dan Kesimpulan):** Statistik adalah angka yang dihasilkan dari sampel, yaitu rata-rata sampel sebesar $\overline{x} = \$28$. Kita tidak mengetahui rata-rata populasi $\mu$ secara pasti tanpa melakukan sensus, tetapi kita menggunakan statistik sampel tersebut untuk mendapatkan nilai perkirannya (aproksimasi).

```python
import numpy as np

# Simulasi pengambilan sampel acak pengunjung taman hiburan
# Anggaplah ini adalah data dari 40 pengunjung (dalam dolar)
np.random.seed(42)
data_pengunjung = np.random.normal(loc=28.0, scale=5.0, size=40).round(2)

print("Data sampel (Pengeluaran 40 Pengunjung):")
print(data_pengunjung)

# Menghitung statistik (rata-rata sampel)
rata_rata_sampel = np.mean(data_pengunjung)
print(f"\nStatistik (Rata-rata Sampel x-bar): ${rata_rata_sampel:.2f}")
```

## 3. Penyajian Data

Dalam statistika deskriptif, data perlu disajikan agar mudah dibaca dan dipahami. Terdapat dua format penyajian data set yang paling umum untuk tahap awal:
*   **Daftar Data (Data List):** Daftar eksplisit dari semua pengukuran individu, baik ditampilkan dengan spasi di antara pengukuran maupun dalam notasi himpunan dengan elemen yang dipisahkan oleh koma.
*   **Tabel Frekuensi Data:** Sebuah tabel di mana setiap nilai unik, disimbolkan dengan $x$, didaftarkan pada baris pertama. Kemudian, frekuensinya, disimbolkan dengan $f$ (yaitu jumlah kemunculan nilai $x$ dalam kumpulan data), didaftarkan tepat di bawahnya pada baris kedua. Tabel ini sangat nyaman digunakan ketika data set berukuran besar namun jumlah nilai uniknya tidak terlalu banyak.

## 4. Contoh Soal & Pembahasan

**Soal:**
Diberikan data umur 21 mahasiswa yang dipilih secara acak yang terdaftar di mata kuliah mahasiswa baru: 
$18, 18, 19, 19, 19, 18, 22, 20, 18, 18, 17, 19, 18, 24, 18, 20, 18, 21, 20, 17, 19$. 
Ubahlah daftar data tersebut ke dalam bentuk tabel frekuensi data!

**Pembahasan:**
*   **Langkah 1 (Identifikasi Nilai Unik $x$):** Urutkan data dari yang terkecil ke yang terbesar untuk menemukan nilai umur yang unik. Nilai unik yang ada adalah 17, 18, 19, 20, 21, 22, 24.
*   **Langkah 2 (Hitung Frekuensi $f$):** Hitung berapa kali setiap angka unik muncul di dalam daftar himpunan data.
    *   Umur 17 muncul 2 kali.
    *   Umur 18 muncul 8 kali.
    *   Umur 19 muncul 5 kali.
    *   Umur 20 muncul 3 kali.
    *   Umur 21 muncul 1 kali.
    *   Umur 22 muncul 1 kali.
    *   Umur 24 muncul 1 kali.
*   **Langkah 3 (Bentuk Tabel):** Masukkan ke dalam tabel dua baris ($x$ dan $f$).

| $x$ | 17 | 18 | 19 | 20 | 21 | 22 | 24 |
|---|---|---|---|---|---|---|---|
| $f$ | 2 | 8 | 5 | 3 | 1 | 1 | 1 |

```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# Data list dari umur mahasiswa
umur_mahasiswa = [18, 18, 19, 19, 19, 18, 22, 20, 18, 18, 17, 19, 18, 24, 18, 20, 18, 21, 20, 17, 19]

# 1. Menghitung frekuensi menggunakan Counter
frekuensi = Counter(umur_mahasiswa)

# 2. Mengubah menjadi DataFrame (Tabel Frekuensi) agar rapi
df_freq = pd.DataFrame.from_dict(frekuensi, orient='index', columns=['f']).reset_index()
df_freq = df_freq.rename(columns={'index': 'x'}).sort_values(by='x')
df_freq = df_freq.set_index('x').T # Transpose agar format baris atas x dan bawah f

print("Tabel Frekuensi Data:")
display(df_freq)

# 3. Visualisasi Data
plt.figure(figsize=(8, 4))
plt.bar(df_freq.columns, df_freq.values[0], color='skyblue', edgecolor='black')
plt.xlabel('Umur (x)')
plt.ylabel('Frekuensi (f)')
plt.title('Distribusi Umur Mahasiswa Baru')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

## 5. Tinjauan Inferensi, Pengambilan Sampel, dan Proporsi

Nilai yang dihitung dari suatu sampel disebut sebagai variabel acak, karena angkanya akan bervariasi dari satu percobaan ke percobaan lainnya dengan cara yang tidak dapat diprediksi secara tepat. Oleh karena itu, kita tidak bisa memastikan nilai eksak dari suatu populasi hanya dari satu sampel, melainkan membuat pernyataan probabilitas.

Kualitas data sangat bergantung pada metode pengambilan sampelnya. Agar data yang diperoleh tidak terdistorsi atau bias, skema pengambilan sampel harus bersifat acak (random) atau representatif terhadap keseluruhan populasi. Salah satu estimasi yang sering dihitung dari sampel adalah proporsi sampel, disimbolkan dengan $\hat{p}$, yang digunakan untuk mengestimasi proporsi populasi $p$.

Rumus untuk proporsi sampel adalah:
$$\hat{p} = \frac{x}{n}$$
Di mana:
*   $x$ = Jumlah kejadian atau elemen dengan karakteristik kualitatif tertentu dalam sampel.
*   $n$ = Total ukuran sampel.

## 6. Contoh Soal & Pembahasan

**Soal:**
Seorang sosiolog ingin memperkirakan proporsi seluruh orang dewasa di suatu wilayah tertentu yang belum pernah menikah. Dalam sebuah sampel acak yang terdiri dari 1.320 orang dewasa, 145 di antaranya belum pernah menikah. Hitunglah statistik yang terlibat dan estimasi proporsi populasi tersebut!

**Pembahasan:**
*   **Langkah 1 (Identifikasi Variabel):**
    Diketahui ukuran sampel $n = 1320$ orang.
    Diketahui jumlah orang yang belum menikah $x = 145$ orang.
*   **Langkah 2 (Menghitung Statistik Proporsi Sampel):**
    Gunakan rumus $\hat{p} = \frac{x}{n}$.
    $$\hat{p} = \frac{145}{1320}$$
    $$\hat{p} \approx 0.11$$
*   **Langkah 3 (Menarik Kesimpulan Inferensial):**
    Statistik yang dihitung dari sampel adalah sekitar $0.11$ atau $11\%$. Berdasarkan sampel tersebut, meskipun kita tidak mengetahui secara pasti persentase proporsi seluruh populasi di wilayah itu, kita dapat mengestimasi bahwa proporsi orang dewasa yang belum pernah menikah adalah sekitar $11\%$.

```python
# Program perhitungan proporsi sampel (p-hat)
n_sampel = 1320    # Total sampel (n)
x_kejadian = 145   # Kasus belum pernah menikah (x)

# Kalkulasi proporsi
p_hat = x_kejadian / n_sampel
persentase = p_hat * 100

print(f"Total Sampel (n) = {n_sampel}")
print(f"Jumlah Kasus (x) = {x_kejadian}")
print(f"Proporsi Sampel (p-hat) = {p_hat:.4f}")
print(f"Estimasi Persentase Populasi = {persentase:.0f}%")
```
