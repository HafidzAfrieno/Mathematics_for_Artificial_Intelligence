## 3.1. Ruang Sampel, Kejadian, dan Probabilitasnya

### Konsep Dasar & Teori
Dalam statistik inferensial, **eksperimen acak** (*random experiment*) adalah suatu proses atau tindakan yang menghasilkan keluaran (*outcome*) pasti, tetapi keluaran tersebut tidak dapat diprediksi dengan pasti sebelum eksperimen dilakukan.

* **Ruang Sampel ($S$)**: Himpunan dari semua keluaran (*outcome*) yang mungkin terjadi dari suatu eksperimen acak.
* **Kejadian (*Event*, $E$)**: Subhimpunan (*subset*) dari ruang sampel $S$. Kejadian $E$ dikatakan terjadi jika keluaran yang diamati merupakan elemen dari himpunan $E$.
* **Probabilitas Kejadian $P(E)$**: Nilai antara $0$ dan $1$ yang mengukur tingkat kepastian/kemungkinan terjadinya suatu kejadian.

Jika semua keluaran dalam ruang sampel terhingga $S = \{e_1, e_2, \dots, e_N\}$ memiliki peluang yang sama (*equally likely outcomes*), maka probabilitas kejadian $E$ didefinisikan sebagai:

$$P(E) = \frac{n(E)}{n(S)}$$

Di mana:
* $n(E)$ adalah jumlah anggota/keluaran dalam kejadian $E$.
* $n(S)$ adalah jumlah total anggota dalam ruang sampel $S$.

Secara umum, probabilitas dari suatu kejadian $E = \{e_1, e_2, \dots, e_k\}$ adalah penjumlahan dari probabilitas masing-masing elemen penyusunnya:

$$P(E) = P(e_1) + P(e_2) + \dots + P(e_k)$$

Aturan dasar probabilitas:
1. $0 \le P(E) \le 1$ untuk setiap kejadian $E$.
2. $P(S) = 1$, jumlah probabilitas seluruh anggota ruang sampel adalah $1$.

```python
import numpy as np
import pandas as pd

# Menghitung Ruang Sampel dan Probabilitas Pelemparan Dua Dadu Fair
dadu1 = np.arange(1, 7)
dadu2 = np.arange(1, 7)

# Membuat ruang sampel (pasangan nilai dadu)
ruang_sampel = [(d1, d2) for d1 in dadu1 for d2 in dadu2]
n_S = len(ruang_sampel)

print(f"Jumlah anggota ruang sampel n(S): {n_S}")

# Kejadian E: Jumlah kedua dadu bernilai genap
kejadian_E = [(d1, d2) for d1, d2 in ruang_sampel if (d1 + d2) % 2 == 0]
n_E = len(kejadian_E)
P_E = n_E / n_S

# Kejadian T: Jumlah kedua dadu lebih dari 8
kejadian_T = [(d1, d2) for d1, d2 in ruang_sampel if (d1 + d2) > 8]
n_T = len(kejadian_T)
P_T = n_T / n_S

print(f"Jumlah kejadian E (Jumlah genap) n(E): {n_E} | P(E) = {P_E:.4f}")
print(f"Jumlah kejadian T (Jumlah > 8) n(T): {n_T} | P(T) = {P_T:.4f}")
```

### Pembahasan Langkah demi Langkah

#### 1. Menentukan Ruang Sampel ($S$)
Dadu pertama memiliki 6 sisi $\{1, 2, 3, 4, 5, 6\}$ dan dadu kedua memiliki 6 sisi. Total keluaran yang mungkin adalah:
$$n(S) = 6 \times 6 = 36$$

#### 2. Menyelesaikan Bagian (1): Peluang Jumlah $\ge 9$
Kejadian $A = \{X \ge 9\}$ terdiri dari jumlah bernilai $9, 10, 11,$ atau $12$:
* Jumlah 9: $(3,6), (4,5), (5,4), (6,3)$ $\rightarrow$ 4 anggota
* Jumlah 10: $(4,6), (5,5), (6,4)$ $\rightarrow$ 3 anggota
* Jumlah 11: $(5,6), (6,5)$ $\rightarrow$ 2 anggota
* Jumlah 12: $(6,6)$ $\rightarrow$ 1 anggota

Total anggota kejadian $A$:
$$n(A) = 4 + 3 + 2 + 1 = 10$$

Maka probabilitasnya adalah:
$$P(X \ge 9) = \frac{n(A)}{n(S)} = \frac{10}{36} \approx 0.2778$$

#### 3. Menyelesaikan Bagian (2): Peluang Jumlah Genap
Kejadian $B = \{X \text{ adalah genap}\}$ terdiri dari jumlah $2, 4, 6, 8, 10, 12$:
* Jumlah 2: $(1,1)$ $\rightarrow$ 1
* Jumlah 4: $(1,3), (2,2), (3,1)$ $\rightarrow$ 3
* Jumlah 6: $(1,5), (2,4), (3,3), (4,2), (5,1)$ $\rightarrow$ 5
* Jumlah 8: $(2,6), (3,5), (4,4), (5,3), (6,2)$ $\rightarrow$ 5
* Jumlah 10: $(4,6), (5,5), (6,4)$ $\rightarrow$ 3
* Jumlah 12: $(6,6)$ $\rightarrow$ 1

Total anggota kejadian $B$:
$$n(B) = 1 + 3 + 5 + 5 + 3 + 1 = 18$$

Maka probabilitasnya adalah:
$$P(X \text{ genap}) = \frac{n(B)}{n(S)} = \frac{18}{36} = 0.5$$

```python
import matplotlib.pyplot as plt
import numpy as np

# Simulasi Distribusi Probabilitas Jumlah Dua Dadu
dadu1 = np.arange(1, 7)
dadu2 = np.arange(1, 7)
jumlah = [d1 + d2 for d1 in dadu1 for d2 in dadu2]

nilai_unik, frekuensi = np.unique(jumlah, return_counts=True)
probabilitas = frekuensi / len(jumlah)

# Visualisasi Distribusi Probabilitas
plt.figure(figsize=(8, 4))
plt.bar(nilai_unik, probabilitas, color='skyblue', edgecolor='black')
plt.xlabel('Jumlah Angka Kedua Dadu (X)')
plt.ylabel('Probabilitas P(X)')
plt.title('Distribusi Probabilitas Pelemparan Dua Dadu')
plt.xticks(nilai_unik)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

## 3.2. Komplementer, Irisan, dan Gabungan Kejadian

### Konsep Dasar & Teori

#### 1. Komplementer (*Complement*)
Komplementer dari kejadian $A$ (dilambangkan $A^c$) adalah himpunan semua keluaran di $S$ yang tidak termasuk dalam $A$.

**Aturan Probabilitas Komplementer:**
$$P(A^c) = 1 - P(A)$$

#### 2. Irisan (*Intersection*)
Irisan dari dua kejadian $A$ dan $B$ (dilambangkan $A \cap B$) adalah himpunan keluaran yang menjadi anggota dari **kedua** kejadian $A$ dan $B$ sekaligus (menggunakan kata hubung "dan").

#### 3. Kejadian Saling Lepas (*Mutually Exclusive*)
Dua kejadian $A$ dan $B$ dikatakan saling lepas jika tidak memiliki anggota yang sama ($A \cap B = \emptyset$). Artinya, kedua kejadian tersebut tidak mungkin terjadi secara bersamaan.

**Aturan untuk Kejadian Saling Lepas:**
$$P(A \cap B) = 0$$

#### 4. Gabungan (*Union*)
Gabungan dari dua kejadian $A$ dan $B$ (dilambangkan $A \cup B$) adalah himpunan keluaran yang menjadi anggota $A$, atau $B$, atau keduanya (menggunakan kata hubung "atau").

**Aturan Penjumlahan Probabilitas (*Additive Rule of Probability*):**
$$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$

Jika $A$ dan $B$ saling lepas, maka:
$$P(A \cup B) = P(A) + P(B)$$

```python
# Simulasi Aturan Penjumlahan Probabilitas (Additive Rule)

# Misal sebuah dadu dilempar
S = set(range(1, 7))

# Kejadian E: Angka Genap
E = {2, 4, 6}
# Kejadian T: Angka Lebih Besar dari 2
T = {3, 4, 5, 6}

# Irisan dan Gabungan
irisan_E_T = E.intersection(T) # E AND T
gabungan_E_T = E.union(T)       # E OR T

P_E = len(E) / len(S)
P_T = len(T) / len(S)
P_irisan = len(irisan_E_T) / len(S)
P_gabungan_hitung = len(gabungan_E_T) / len(S)

# Menggunakan Rumus Additive Rule
P_gabungan_rumus = P_E + P_T - P_irisan

print(f"E = {E}, P(E) = {P_E:.4f}")
print(f"T = {T}, P(T) = {P_T:.4f}")
print(f"E ∩ T = {irisan_E_T}, P(E ∩ T) = {P_irisan:.4f}")
print(f"P(E ∪ T) dari hitungan langsung = {P_gabungan_hitung:.4f}")
print(f"P(E ∪ T) dari rumus (P(E)+P(T)-P(E∩T)) = {P_gabungan_rumus:.4f}")
```

## 3.3. Probabilitas Bersyarat dan Kejadian Saling Bebas

### Konsep Dasar & Teori

#### 1. Probabilitas Bersyarat (*Conditional Probability*)
Probabilitas bersyarat $P(A|B)$ adalah peluang terjadinya kejadian $A$ dengan syarat/pengetahuan bahwa kejadian $B$ **telah terjadi**.

**Rumus Probabilitas Bersyarat:**
$$P(A|B) = \frac{P(A \cap B)}{P(B)}, \quad \text{dengan } P(B) > 0$$

Atau sebaliknya:
$$P(B|A) = \frac{P(A \cap B)}{P(A)}, \quad \text{dengan } P(A) > 0$$

Dari rumus ini, kita memperoleh **Aturan Perkalian Probabilitas**:
$$P(A \cap B) = P(B) \cdot P(A|B) = P(A) \cdot P(B|A)$$

#### 2. Kejadian Saling Bebas (*Independent Events*)
Dua kejadian $A$ dan $B$ dikatakan **saling bebas** (*independent*) jika terjadinya kejadian $B$ tidak mempengaruhi peluang terjadinya kejadian $A$, dan sebaliknya.

**Syarat Kejadian Saling Bebas:**
1. $P(A|B) = P(A)$
2. $P(B|A) = P(B)$
3. $P(A \cap B) = P(A) \cdot P(B)$

Jika syarat di atas tidak terpenuhi, maka $A$ dan $B$ dikatakan **saling bergantung** (*dependent*).

```python
# Perhitungan Probabilitas Bersyarat dan Pengujian Independensi

# Data dari Tabel Kontingensi (Misal: Kasus Hipertensi vs Berat Badan)
# Baris: Overweight (O), Not Overweight (O')
# Kolom: Hypertension (H), No Hypertension (H')

# Probabilitas Gabungan
P_H_and_O = 0.09
P_Hc_and_O = 0.02
P_H_and_Oc = 0.11
P_Hc_and_Oc = 0.78

# Total Marginal
P_O = P_H_and_O + P_Hc_and_O      # Total Overweight = 0.11
P_Oc = P_H_and_Oc + P_Hc_and_Oc   # Total Not Overweight = 0.89
P_H = P_H_and_O + P_H_and_Oc      # Total Hypertension = 0.20

# 1. Menhitung P(H|O) -> Peluang Hipertensi JIKA Overweight
P_H_given_O = P_H_and_O / P_O

# 2. Menhitung P(H|O') -> Peluang Hipertensi JIKA TIDAK Overweight
P_H_given_Oc = P_H_and_Oc / P_Oc

# 3. Uji Independensi
is_independent = np.isclose(P_H_and_O, P_H * P_O)

print(f"P(H|O)  = {P_H_given_O:.4f}")
print(f"P(H|O') = {P_H_given_Oc:.4f}")
print(f"Apakah Kejadian Hipertensi & Overweight Saling Bebas? {is_independent}")
```

### Soal 3
Sebuah toples berisi 10 kelereng: 7 berwarna hitam ($B$) dan 3 berwarna putih ($W$). Dua kelereng diambil satu per satu **tanpa pengembalian** (*without replacement*).

1. Hitung peluang bahwa kedua kelereng yang terambil berwarna hitam ($P(B_1 \cap B_2)$).
2. Hitung peluang bahwa kelereng yang terambil tepat satu berwarna hitam.
3. Hitung peluang bahwa setidaknya satu kelereng berwarna hitam.


### Pembahasan Langkah demi Langkah

#### 1. Pengambilan Pertama
* Total kelereng = $10$ (7 Hitam, 3 Putih)
* Peluang terambil kelereng hitam pertama:
$$P(B_1) = \frac{7}{10} = 0.7$$
* Peluang terambil kelereng putih pertama:
$$P(W_1) = \frac{3}{10} = 0.3$$

#### 2. Pengambilan Kedua (Tanpa Pengembalian)
Sisa kelereng dalam toples menjadi $9$ buah.

* **Kondisi A (Jika kelereng 1 Hitam):** Sisa kelereng = 6 Hitam, 3 Putih.
  $$P(B_2 | B_1) = \frac{6}{9}, \quad P(W_2 | B_1) = \frac{3}{9}$$
* **Kondisi B (Jika kelereng 1 Putih):** Sisa kelereng = 7 Hitam, 2 Putih.
  $$P(B_2 | W_1) = \frac{7}{9}, \quad P(W_2 | W_1) = \frac{2}{9}$$

#### 3. Menyelesaikan Bagian (1): Kedua Kelereng Hitam ($B_1 \cap B_2$)
$$P(B_1 \cap B_2) = P(B_1) \cdot P(B_2 | B_1) = \left(\frac{7}{10}\right) \cdot \left(\frac{6}{9}\right) = \frac{42}{90} \approx 0.4667$$

#### 4. Menyelesaikan Bagian (2): Tepat Satu Kelereng Hitam
Kombinasi kejadian: $(B_1 \cap W_2)$ atau $(W_1 \cap B_2)$. Kedua kejadian ini saling lepas.

$$P(\text{Tepat 1 Hitam}) = P(B_1 \cap W_2) + P(W_1 \cap B_2)$$
$$P(B_1 \cap W_2) = \left(\frac{7}{10}\right) \cdot \left(\frac{3}{9}\right) = \frac{21}{90}$$
$$P(W_1 \cap B_2) = \left(\frac{3}{10}\right) \cdot \left(\frac{7}{9}\right) = \frac{21}{90}$$
$$P(\text{Tepat 1 Hitam}) = \frac{21}{90} + \frac{21}{90} = \frac{42}{90} \approx 0.4667$$

#### 5. Menyelesaikan Bagian (3): Setidaknya Satu Kelereng Hitam
Dapat dihitung dengan aturan komplementer dari kejadian "keduanya putih" ($W_1 \cap W_2$):
$$P(W_1 \cap W_2) = \left(\frac{3}{10}\right) \cdot \left(\frac{2}{9}\right) = \frac{6}{90} \approx 0.0667$$

$$P(\text{Setidaknya 1 Hitam}) = 1 - P(W_1 \cap W_2) = 1 - \frac{6}{90} = \frac{84}{90} \approx 0.9333$$

```python
# Visualisasi Pohon Keputusan (Tree Diagram) Probabilitas

import matplotlib.pyplot as plt

# Nilai Peluang
p_B1 = 7/10
p_W1 = 3/10

p_B2_given_B1 = 6/9
p_W2_given_B1 = 3/9

p_B2_given_W1 = 7/9
p_W2_given_W1 = 2/9

# Hasil Akhir
p_BB = p_B1 * p_B2_given_B1
p_BW = p_B1 * p_W2_given_B1
p_WB = p_W1 * p_B2_given_W1
p_WW = p_W1 * p_W2_given_W1

print(f"P(BB) = {p_BB:.4f}")
print(f"P(BW) = {p_BW:.4f}")
print(f"P(WB) = {p_WB:.4f}")
print(f"P(WW) = {p_WW:.4f}")
print(f"Total Probabilitas = {p_BB + p_BW + p_WB + p_WW:.1f}")
```

**7.** Sebuah gudang menerima pasokan komponen elektronik dari tiga vendor berbeda: Vendor I menyuplai 50%, Vendor II menyuplai 30%, dan Vendor III menyuplai 20%. Persentase komponen cacat yang dihasilkan oleh Vendor I, II, dan III masing-masing adalah 3%, 5%, dan 8%.

```python
# a. Berapa probabilitas satu komponen yang diambil secara acak dari gudang tersebut cacat? 
vendor_sumpllier = [0.5,0.3,0.2]
rusak_vendor     = [0.03,0.05,0.08]
total_kerusakan_vendor = 0

for i in range(0,3):
    total_kerusakan_vendor += (vendor_sumpllier[i] * rusak_vendor[i])
print(f' probabilitas satu komponen cacat di gudang adalah: {total_kerusakan_vendor}')
```

```python
# b. Jika komponen yang diambil secara acak ternyata cacat, berapa probabilitas komponen tersebut dikirim oleh Vendor III?
komponen_cacat_vendor_b = (vendor_sumpllier[2] * rusak_vendor[2]) / total_kerusakan_vendor
print(f'Pelunang komponen cacat dari supplier B adalah {komponen_cacat_vendor_b:.4f}')
```

**8.** Kotak kayu berisi 5 bola merah, 3 bola hijau, dan 2 bola putih. Jika dua bola diambil satu per satu secara berurutan tanpa pengembalian: 

```python
# a. Berapa probabilitas bola pertama berwarna merah dan bola kedua berwarna hijau? 
bola_merah,bola_hijau,bola_putih = 5,3,2
total_bola = bola_merah + bola_hijau + bola_putih

total_ambil_bola = math.perm(total_bola,2)
ambil_merah = math.perm(bola_merah,1)
ambil_hijau = math.perm(bola_hijau,1)
ambil_hijau_dan_merah = ambil_merah * ambil_hijau

peluang_bola_hijaumerah = ambil_hijau_dan_merah / total_ambil_bola
print(f' probabilitas bola pertama berwarna merah dan bola kedua berwarna hijau adalah {peluang_bola_hijaumerah:.4f}')
```

```python
# b. Berapa probabilitas kedua bola yang terambil memiliki warna yang sama?
ambil_merah_sama = math.perm(bola_merah,2)
ambil_hijau_sama = math.perm(bola_hijau,2)
ambil_putih_sama = math.perm(bola_putih,2)

totalcara_bolasama = ambil_merah_sama + ambil_hijau_sama + ambil_putih_sama
peluang_ketiga_bola = totalcara_bolasama / total_ambil_bola

print(f'probabilitas kedua bola yang terambil memiliki warna yang sama adalah {peluang_ketiga_bola:.4f}')

```

**9.** Suatu tes medis digunakan untuk mendiagnosis suatu penyakit langka yang menginfeksi 1%dari populasi. Tes tersebut memiliki akurasi 95%untuk mendeteksi orang yang benar-benar sakit (positif benar) dan memberikan hasil positif palsu sebesar 5%pada orang yang sehat. Jika seorang pasien dites dan hasilnya positif, berapa probabilitas sebenarnya pasien tersebut benar-benar menderita penyakit tersebut?

```python
terinfeksi = 0.01
positif_benar = 0.95
posiitf_palsu = 1 - positif_benar
non_terinfeksi = 1 - terinfeksi

total_peluang_pasien_positif = (terinfeksi * positif_benar) + (non_terinfeksi * posiitf_palsu)
peluang_real_sakit = (terinfeksi * positif_benar) / total_peluang_pasien_positif

print(f'Peluang pasien benar-benar sakit adalah {peluang_real_sakit:.4f}')
```

**10.** Dalam sebuah kompetisi e-sports, Tim Alpha memiliki probabilitas menang 0,60jika bertanding saat cuaca cerah dan 0,35jika bertanding saat hujan. Berdasarkan prakiraan cuaca, peluang terjadinya hujan pada hari pertandingan adalah 0,40. 

```python
# a. Hitung probabilitas total Tim Alpha menang dalam pertandingan tersebut. 
p_hujan = 0.4
p_tidak_hujan = 1 - p_hujan
p_menang_hujan = 0.35
p_menang_tdkhujan = 0.6

peluang_menang = (p_hujan * p_menang_hujan) + (p_tidak_hujan * p_menang_tdkhujan)

print(f'probabilitas total Tim Alpha menang dalam pertandingan tersebut adalah {peluang_menang:.4f}')
```

```python
peluang_hujanmenang = (p_hujan * p_menang_hujan) / peluang_menang

print(f'Peluang Tim Alpha menang ketika hujan adalah {peluang_hujanmenang:.4f}')
```
