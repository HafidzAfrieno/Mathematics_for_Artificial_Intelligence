## 4.1. Variabel Acak (Random Variables)

Variabel acak adalah suatu kuantitas numerik yang nilainya dihasilkan dari sebuah eksperimen acak. Variabel acak secara umum disimbolkan dengan huruf kapital seperti $X$ atau $Z$, sedangkan nilai spesifik/riil yang dapat diambil oleh variabel tersebut disimbolkan dengan huruf kecil seperti $x$ atau $z$.

Variabel acak diklasifikasikan menjadi dua jenis utama berdasarkan sifat nilai yang mungkin diambilnya:

1. **Variabel Acak Diskrit (Discrete Random Variable)**: Variabel acak yang memiliki jumlah nilai mungkin yang terhingga (*finite*) atau dapat dihitung (*countable*). Variabel acak diskrit umumnya muncul dari proses pemetaan pencacahan atau perhitungan (misalnya: jumlah kejadian, jumlah barang rusak, jumlah pelemparan koin).
2. **Variabel Acak Kontinu (Continuous Random Variable)**: Variabel acak yang nilai-nilai mungkinkah membentuk suatu interval bilangan riil. Variabel acak kontinu umumnya muncul dari proses pengukuran (misalnya: tinggi badan, berat, waktu, suhu).


## 4.2. Distribusi Probabilitas, Nilai Harapan (Mean), dan Varians Diskrit

### A. Fungsi Distribusi Probabilitas (Probability Distribution)
Distribusi probabilitas dari variabel acak diskrit $X$ adalah daftar yang memuat seluruh nilai yang mungkin dari $X$ beserta nilai probabilitas $P(x)$ yang bersesuaian dengan setiap nilai $x$ tersebut.

Setiap fungsi probabilitas $P(x)$ dari variabel acak diskrit harus memenuhi dua syarat utama:
1. Nilai probabilitas untuk setiap kejadian $x$ berada di rentang 0 hingga 1:
$$0 \le P(x) \le 1$$
2. Total jumlah seluruh probabilitas dari semua nilai yang mungkin adalah 1:
$$\sum P(x) = 1$$

### B. Nilai Harapan / Rata-rata ($\mu$ atau $E(X)$)
Nilai harapan (ekspektasi/mean) dari variabel acak diskrit $X$ mewakili rata-rata jangka panjang dari nilai variabel acak jika eksperimen dilakukan secara berulang-ulang.
$$\mu = E(X) = \sum x \cdot P(x)$$

### C. Varians ($\sigma^2$) dan Deviasi Standar ($\sigma$)
Varians mengukur seberapa jauh persebaran nilai-nilai variabel acak dari nilai harapannya ($\mu$).

Formula konseptual varians:
$$\sigma^2 = \sum (x - \mu)^2 \cdot P(x)$$

Formula komputasi (shortcut):
$$\sigma^2 = \left[ \sum x^2 \cdot P(x) \right] - \mu^2$$

Deviasi standar ($\sigma$) adalah akar kuadrat dari nilai varians:
$$\sigma = \sqrt{\sigma^2} = \sqrt{\left[ \sum x^2 \cdot P(x) \right] - \mu^2}$$


## 4.3. Distribusi Binomial (The Binomial Distribution)

Eksperimen Binomial adalah eksperimen acak yang memenuhi empat kriteria khusus:
1. Terdiri dari $n$ percobaan yang identik dan independen.
2. Setiap percobaan hanya memiliki dua kemungkinan hasil: **"Sukses"** (dinotasikan $S$) atau **"Gagal"** (dinotasikan $F$).
3. Probabilitas "Sukses" pada setiap percobaan adalah konstan, dilambangkan dengan $p$. Probabilitas "Gagal" adalah $q = 1 - p$.
4. Variabel acak $X$ menghitung jumlah total "Sukses" dalam $n$ percobaan tersebut.

### Formula Probabilitas Binomial
Probabilitas mendapatkan tepat $x$ sukses dalam $n$ percobaan binomial dirumuskan dengan:
$$P(x) = \binom{n}{x} p^x q^{n-x} = \frac{n!}{x!(n - x)!} p^x (1 - p)^{n-x}$$

Di mana $x = 0, 1, 2, \dots, n$ dan $n! = 1 \cdot 2 \cdot 3 \cdots m$ (dengan $0! = 1$).

### Rata-rata dan Deviasi Standar Khusus Distribusi Binomial
Khusus untuk variabel acak binomial, perhitungan nilai rata-rata, varians, dan deviasi standar dapat disederhanakan menjadi:
* **Mean ($\mu$)**: 
  $$\mu = n \cdot p$$
* **Varians ($\sigma^2$)**: 
  $$\sigma^2 = n \cdot p \cdot q = n \cdot p \cdot (1 - p)$$
* **Deviasi Standar ($\sigma$)**: 
  $$\sigma = \sqrt{n \cdot p \cdot q} = \sqrt{n \cdot p \cdot (1 - p)}$$

```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Definisi Data Distribusi Probabilitas Undian
x_values = np.array([299, 199, 99, -1])
p_values = np.array([0.001, 0.001, 0.001, 0.997])

# 2. Perhitungan Statistik Dasar (Mean, Varians, Deviasi Standar)
mean_X = np.sum(x_values * p_values)
var_X = np.sum(((x_values - mean_X) ** 2) * p_values)
std_X = np.sqrt(var_X)

print("=== HASIL PERHITUNGAN STATISTIK DISKRIT ===")
print(f"E(X) / Mean (mu)          : {mean_X:.2f}")
print(f"Varians (sigma^2)         : {var_X:.2f}")
print(f"Deviasi Standar (sigma)   : {std_X:.4f}")

# 3. Visualisasi Fungsi Probabilitas (PMF)
plt.figure(figsize=(8, 5))
markerline, stemlines, baseline = plt.stem(
    x_values,
    p_values,
    linefmt="b-",
    markerfmt="bo",
    basefmt="r-",
)
plt.setp(markerline, markersize=8)

plt.title(
    "Distribusi Probabilitas Keuntungan Bersih Undian (PMF)",
    fontsize=12,
    fontweight="bold",
)
plt.xlabel("Keuntungan Bersih / Net Gain ($)", fontsize=10)
plt.ylabel("Probabilitas P(x)", fontsize=10)
plt.grid(True, linestyle="--", alpha=0.6)

# Menambahkan label nilai probabilitas di atas batang
for x, p in zip(x_values, p_values):
    plt.text(
        x,
        p + 0.03,
        f"{p:.3f}",
        ha="center",
        va="bottom",
        fontsize=9,
        fontweight="bold",
    )

plt.ylim(0, 1.1)
plt.show()
```

## 4.4. Contoh Soal & Pembahasan

### Contoh Soal 1: Distribusi Probabilitas Diskrit & Nilai Harapan
Sebuah organisasi mengadakan undian kupon. Sebanyak 1.000 kupon dijual seharga \$1 per kupon. Terdapat satu hadiah utama sebesar \$300, satu hadiah kedua sebesar \$200, dan satu hadiah ketiga sebesar \$100. Misalkan $X$ menyatakan keuntungan bersih (*net gain*) dari pembelian satu kupon undian.

**Pertanyaan:**
1. Buatlah tabel distribusi probabilitas untuk $X$.
2. Hitung nilai harapan $E(X)$ dan interpretasikan hasilnya.
3. Hitung varians ($\sigma^2$) dan deviasi standar ($\sigma$) dari $X$.

#### **Pembahasan Langkah demi Langkah:**

**Langkah 1: Menentukan nilai $X$ dan $P(x)$**
Keuntungan bersih dihitung dari (Nilai Hadiah - Harga Kupon \$1):
* Menang Hadiah I: $x = 300 - 1 = 299$, dengan probabilitas $P(299) = \frac{1}{1000} = 0.001$
* Menang Hadiah II: $x = 200 - 1 = 199$, dengan probabilitas $P(199) = \frac{1}{1000} = 0.001$
* Menang Hadiah III: $x = 100 - 1 = 99$, dengan probabilitas $P(99) = \frac{1}{1000} = 0.001$
* Kalah (tidak mendapat hadiah): $x = 0 - 1 = -1$, dengan probabilitas $P(-1) = \frac{997}{1000} = 0.997$

Tabel Distribusi Probabilitas $X$:
| $x$ | $P(x)$ |
| :---: | :---: |
| 299 | 0.001 |
| 199 | 0.001 |
| 99 | 0.001 |
| -1 | 0.997 |


**Langkah 2: Menghitung Nilai Harapan $E(X)$**
$$E(X) = \sum x \cdot P(x)$$
$$E(X) = (299 \cdot 0.001) + (199 \cdot 0.001) + (99 \cdot 0.001) + ((-1) \cdot 0.997)$$
$$E(X) = 0.299 + 0.199 + 0.099 - 0.997 = -0.40$$

**Interpretasi**: Nilai negatif menyatakan kerugian rata-rata. Pembeli kupon secara rata-rata diperkirakan akan rugi sebesar \$0.40 (atau 40 sen) per kupon yang dibeli dalam jangka panjang.


**Langkah 3: Menghitung Varians ($\sigma^2$) dan Deviasi Standar ($\sigma$)**
Gunakan rumus $\sigma^2 = \left[ \sum x^2 \cdot P(x) \right] - \mu^2$:

1. Hitung $\sum x^2 \cdot P(x)$:
   $$\sum x^2 \cdot P(x) = (299^2 \cdot 0.001) + (199^2 \cdot 0.001) + (99^2 \cdot 0.001) + ((-1)^2 \cdot 0.997)$$
   $$\sum x^2 \cdot P(x) = (89401 \cdot 0.001) + (39601 \cdot 0.001) + (9801 \cdot 0.001) + (1 \cdot 0.997)$$
   $$\sum x^2 \cdot P(x) = 89.401 + 39.601 + 9.801 + 0.997 = 139.80$$

2. Hitung Varians ($\sigma^2$):
   $$\sigma^2 = 139.80 - (-0.40)^2 = 139.80 - 0.16 = 139.64$$

3. Hitung Deviasi Standar ($\sigma$):
   $$\sigma = \sqrt{139.64} \approx 11.817$$


### Contoh Soal 2: Penerapan Distribusi Binomial
Menurut data riset, $17\%$ korban penipuan keuangan mengenal pelaku penipuan secara pribadi. Sebuah tim penyidik mengambil sampel acak sebanyak $n = 5$ kasus penipuan keuangan secara independen. Misalkan $X$ menyatakan jumlah kasus di mana korban mengenal pelakunya secara pribadi.

**Pertanyaan:**
1. Hitunglah probabilitas bahwa tepat 2 korban mengenal pelakunya secara pribadi ($P(X = 2)$).
2. Hitung probabilitas bahwa setidaknya 1 korban mengenal pelakunya ($P(X \ge 1)$).
3. Hitung rata-rata ($\mu$) dan deviasi standar ($\sigma$) dari variabel acak binomial $X$.

#### **Pembahasan Langkah demi Langkah:**

**Diketahui**:
* Parameter Binomial: $n = 5$, $p = 0.17$, $q = 1 - 0.17 = 0.83$


**Langkah 1: Menghitung $P(X = 2)$**
Menggunakan rumus probabilitas Binomial:
$$P(x) = \binom{n}{x} p^x q^{n-x}$$
$$P(2) = \binom{5}{2} (0.17)^2 (0.83)^{5-2}$$
$$\binom{5}{2} = \frac{5!}{2!(5-2)!} = \frac{120}{2 \cdot 6} = 10$$
$$P(2) = 10 \cdot (0.0289) \cdot (0.83)^3$$
$$P(2) = 10 \cdot 0.0289 \cdot 0.571787 \approx 0.1652 \text{ (atau } 16.52\%\text{)}$$


**Langkah 2: Menghitung $P(X \ge 1)$**
Menggunakan aturan komplementer probabilitas agar kalkulasi lebih efisien:
$$P(X \ge 1) = 1 - P(X = 0)$$

Hitung $P(X = 0)$:
$$P(0) = \binom{5}{0} (0.17)^0 (0.83)^5 = 1 \cdot 1 \cdot 0.393904 \approx 0.3939$$

Maka:
$$P(X \ge 1) = 1 - 0.393904 = 0.6061 \text{ (atau } 60.61\%\text{)}$$


**Langkah 3: Menghitung Rata-rata ($\mu$) dan Deviasi Standar ($\sigma$)**
1. Rata-rata ($\mu$):
   $$\mu = n \cdot p = 5 \cdot 0.17 = 0.85$$
2. Deviasi Standar ($\sigma$):
   $$\sigma = \sqrt{n \cdot p \cdot q} = \sqrt{5 \cdot 0.17 \cdot 0.83} = \sqrt{0.7055} \approx 0.8399$$

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import binom

# 1. Parameter Distribusi Binomial
n = 5
p = 0.17
q = 1 - p

# 2. Pembuatan Ruang Sampel x (0 sampai n)
x = np.arange(0, n + 1)

# 3. Perhitungan PMF (Probability Mass Function) dan CDF (Cumulative Distribution Function)
pmf = binom.pmf(x, n, p)
cdf = binom.cdf(x, n, p)

# Perhitungan probabilitas spesifik
p_exact_2 = binom.pmf(2, n, p)
p_ge_1 = 1 - binom.pmf(0, n, p)  # atau 1 - binom.cdf(0, n, p)

# Perhitungan Mean dan Deviasi Standar Binomial
mean_binom = binom.mean(n, p)
std_binom = binom.std(n, p)

print("=== HASIL PERHITUNGAN DISTRIBUSI BINOMIAL ===")
print(f"P(X = 2)                   : {p_exact_2:.4f}")
print(f"P(X >= 1)                  : {p_ge_1:.4f}")
print(f"Mean / Rata-rata (mu)      : {mean_binom:.2f}")
print(f"Deviasi Standar (sigma)    : {std_binom:.4f}\n")

# 4. Tabel Distribusi Probabilitas
print("Tabel PMF & CDF Binomial (n=5, p=0.17):")
print("x\tP(X = x)\tP(X <= x)")
print("-" * 30)
for i in range(len(x)):
    print(f"{x[i]}\t{pmf[i]:.4f}\t\t{cdf[i]:.4f}")

# 5. Visualisasi Bar Chart Distribusi Binomial
plt.figure(figsize=(9, 5))
bars = plt.bar(
    x,
    pmf,
    color="skyblue",
    edgecolor="navy",
    alpha=0.8,
    width=0.5,
    label="P(X = x)",
)

# Sorot batang X = 2
bars[2].set_color("orange")
bars[2].set_edgecolor("red")

plt.title(
    f"Distribusi Binomial (n={n}, p={p})", fontsize=12, fontweight="bold"
)
plt.xlabel("Jumlah Korban Mengenal Pelaku (x)", fontsize=10)
plt.ylabel("Probabilitas", fontsize=10)
plt.xticks(x)
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Label nilai probabilitas di atas batang
for bar in bars:
    yval = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2.0,
        yval + 0.01,
        f"{yval:.4f}",
        ha="center",
        va="bottom",
        fontsize=9,
    )

plt.ylim(0, max(pmf) + 0.08)
plt.show()
```
