# **Bab 1: Pengantar Statistik**

## 1. Definisi dan Konsep Dasar
*   **Populasi**: Kumpulan seluruh objek atau individu yang menjadi perhatian dalam suatu studi statistik.
*   **Sampel**: Subset atau subkoleksi dari populasi yang dipilih untuk dianalisis.
*   **Sensus**: Suatu sampel yang mencakup seluruh populasi.
*   **Pengukuran**: Angka atau atribut yang dihitung untuk setiap anggota populasi atau sampel. Kumpulan pengukuran dari elemen sampel disebut **data sampel**.
*   **Parameter**: Angka yang merangkum suatu aspek dari populasi secara keseluruhan. Dilambangkan dengan huruf Yunani, seperti $\mu$ (rata-rata populasi), $\sigma$ (standar deviasi populasi), dan $p$ (proporsi populasi).
*   **Statistik**: Angka yang dihitung dari data sampel. Dilambangkan dengan huruf Latin, seperti $\bar{x}$ (rata-rata sampel), $s$ (standar deviasi sampel), dan $\hat{p}$ (proporsi sampel).
*   **Statistik Deskriptif**: Cabang statistik yang melibatkan pengorganisasian, penyajian, dan pendeskripsian data.
*   **Statistik Inferensial**: Cabang statistik yang melibatkan penarikan kesimpulan tentang populasi berdasarkan informasi yang terkandung dalam sampel.
*   **Data Kualitatif**: Pengukuran yang tidak memiliki skala numerik alami, terdiri dari atribut, label, atau karakteristik non-numerik. Contoh: warna mobil, jenis kelamin.
*   **Data Kuantitatif**: Pengukuran numerik yang berasal dari skala numerik alami. Contoh: nilai mobil, usia.

```python
import numpy as np
import pandas as pd
from collections import Counter

print("=== 1. DEFINISI DAN KONSEP DASAR ===")

# Simulasi Populasi: Nilai ujian (skala 0-100) dari 1000 siswa
np.random.seed(42)
populasi_nilai = np.random.normal(loc=75, scale=10, size=1000) # Rata-rata 75, std 10

# Parameter Populasi
mu = np.mean(populasi_nilai)
sigma = np.std(populasi_nilai)

# Ambil Sampel Acak (n=30)
n = 30
sampel_nilai = np.random.choice(populasi_nilai, size=n, replace=False)

# Statistik Sampel
x_bar = np.mean(sampel_nilai)
s = np.std(sampel_nilai, ddof=1) # ddof=1 untuk standar deviasi sampel

print(f"Rata-rata Populasi (Parameter mu)   : {mu:.2f}")
print(f"Std Dev Populasi (Parameter sigma)  : {sigma:.2f}")
print(f"Rata-rata Sampel (Statistik x_bar)  : {x_bar:.2f}")
print(f"Std Dev Sampel (Statistik s)        : {s:.2f}")

# Contoh Data Kualitatif vs Kuantitatif
data_kualitatif = ["Merah", "Biru", "Hitam", "Putih", "Merah"]
data_kuantitatif = [80, 85, 90, 75, 88]

print(f"\nContoh Data Kualitatif : {data_kualitatif}")
print(f"Contoh Data Kuantitatif: {data_kuantitatif}\n")
```

## 2. Gambaran Umum (Overview)
*   **Variabel Acak**: Nilai statistik sampel (seperti $\bar{x}$) bervariasi dari sampel ke sampel. Oleh karena itu, $\bar{x}$ disebut variabel acak.
*   **Interval Kepercayaan (Confidence Interval)**: Daripada hanya melaporkan satu angka estimasi (*point estimate*), kita dapat membangun interval nilai yang memiliki peluang tertentu (misalnya 95%) untuk memuat parameter populasi sebenarnya. Contoh: Kita 95% yakin rata-rata sebenarnya berada dalam rentang $\bar{x} \pm E$, di mana $E$ adalah *margin of error*.
*   **Pengujian Hipotesis**: Proses pengambilan sampel untuk menguji klaim atau asumsi tentang parameter populasi. Contoh: Menguji apakah rata-rata gula dalam makanan kemasan benar-benar 14 gram atau lebih.
*   **Pentingnya Pengambilan Sampel Acak**: Nilai data hanya sebagus sampel yang menghasilkannya. Sampel yang tidak acak (bias) dapat menghasilkan estimasi yang menyimpang. Dalam kursus ini, kita mengasumsikan sampel bersifat acak atau representatif.

```python
print("=== 2. GAMBARAN UMUM (OVERVIEW) ===")

# Simulasi Variabel Acak: Mengambil 5 sampel berbeda dan melihat rata-ratanya
print("Simulasi Variabilitas Sampel (Variabel Acak):")
for i in range(1, 6):
    sampel_iterasi = np.random.choice(populasi_nilai, size=30, replace=False)
    print(f"  Sampel ke-{i} memiliki rata-rata: {np.mean(sampel_iterasi):.2f}")

# Simulasi Interval Kepercayaan 95% (menggunakan pendekatan Z)
# Margin of Error (E) = z_alpha/2 * (s / sqrt(n)), dengan z_0.025 = 1.96
margin_of_error = 1.96 * (s / np.sqrt(n))
ci_lower = x_bar - margin_of_error
ci_upper = x_bar + margin_of_error
print(f"\nInterval Kepercayaan 95% untuk rata-rata populasi:")
print(f"  {ci_lower:.2f} < mu < {ci_upper:.2f}")
print(f"  (Apakah interval ini memuat mu sebenarnya? {ci_lower <= mu <= ci_upper})\n")
```

## 3. Penyajian Data
*   **Daftar Data (Data List)**: Penyajian data dengan mencantumkan semua pengukuran individu secara eksplisit, baik dipisahkan spasi maupun dalam notasi himpunan.
*   **Tabel Frekuensi Data (Data Frequency Table)**: Tabel di mana setiap nilai berbeda $x$ dicantumkan di baris pertama dan frekuensinya $f$ (jumlah kemunculan nilai $x$) dicantumkan di bawahnya.
*   **Rata-rata dari Tabel Frekuensi**: Jika data disajikan dalam tabel frekuensi, rata-rata sampel dapat dihitung dengan rumus:
    $$\bar{x} = \frac{\sum x f}{n}$$
    di mana $n = \sum f$ adalah ukuran sampel total.

```python
print("=== 3. PENYAJIAN DATA ===")

# Daftar Data
data_list = [2, 3, 3, 4, 4, 4, 5, 5, 5, 5, 6, 6, 7]
print(f"Daftar Data: {data_list}")

# Tabel Frekuensi Data
frekuensi = Counter(data_list)
tabel_frekuensi = pd.DataFrame(list(frekuensi.items()), columns=['Nilai (x)', 'Frekuensi (f)'])
tabel_frekuensi = tabel_frekuensi.sort_values(by='Nilai (x)').reset_index(drop=True)
print("\nTabel Frekuensi Data:")
print(tabel_frekuensi.to_string(index=False))

# Menghitung Rata-rata dari Tabel Frekuensi
total_f = tabel_frekuensi['Frekuensi (f)'].sum()
total_xf = (tabel_frekuensi['Nilai (x)'] * tabel_frekuensi['Frekuensi (f)']).sum()
x_bar_freq = total_xf / total_f
print(f"\nTotal n (sum f) = {total_f}")
print(f"Total sum(xf)   = {total_xf}")
print(f"Rata-rata dari tabel frekuensi (x_bar) = {x_bar_freq:.2f}\n")
```
