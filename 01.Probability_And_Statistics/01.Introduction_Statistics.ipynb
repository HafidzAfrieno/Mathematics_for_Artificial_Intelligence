## 1.1. Basic Definitions and Concepts
Dalam studi statistik, terdapat beberapa konsep dan terminologi dasar yang menjadi landasan utama:
- **Populasi (Population)**: Keseluruhan himpunan atau koleksi objek/individu yang menjadi fokus perhatian dalam suatu penelitian statistik.
- **Sampel (Sample)**: Sebagian kecil atau subsit dari populasi yang dipilih untuk diobservasi. Jika sampel mencakup seluruh anggota populasi, proses ini disebut **Sensus (Census)**.
- **Pengukuran (Measurement)**: Nilai atau atribut yang diukur dari setiap anggota populasi maupun sampel. Data pengukuran pada sampel disebut **Data Sampel (Sample Data)**.
- **Parameter**: Nilai numerik yang merangkum atau mendeskripsikan karakteristik dari seluruh populasi. Nilai parameter umumnya tidak diketahui secara pasti kecuali dilakukan sensus.
- **Statistik (Statistic)**: Nilai numerik yang dihitung berdasarkan data sampel dan digunakan untuk mengestimasi parameter populasi.

Statistika dibagi menjadi dua cabang utama:
1. **Statistika Deskriptif (Descriptive Statistics)**: Metode yang berfokus pada pengorganisasian, penyajian, dan penggambaran data.
2. **Statistika Inferensial (Inferential Statistics)**: Metode yang berfokus pada penarikan kesimpulan atau inferensi mengenai populasi berdasarkan data yang diperoleh dari sampel.

Jenis Data berdasarkan sifatnya:
- **Data Kualitatif (Qualitative Data)**: Data yang berbentuk atribut, label, atau karakteristik non-numerik.
- **Data Kuantitatif (Quantitative Data)**: Data numerik yang diperoleh dari skala pengukuran numerik.

Simbol-simbol matematika dasar yang digunakan:
- Ukuran sampel dinyatakan dengan $n$.
- Proporsi populasi dinyatakan dengan $p$, sedangkan proporsi sampel dinyatakan dengan $\hat{p}$.
- Rata-rata populasi dinyatakan dengan $\mu$, sedangkan rata-rata sampel dinyatakan dengan $\bar{x}$.

## 1.2. Contoh Soal & Pembahasan
**Soal:**
Seorang peneliti ingin mengestimasi proporsi orang dewasa yang memiliki ponsel di suatu daerah. Peneliti mengambil sampel acak sebanyak $1.572$ orang dewasa, dan menemukan bahwa $1.298$ orang di antaranya memiliki ponsel.
1. Tentukan populasi dan parameter yang diteliti!
2. Hitung statistik sampelnya (proporsi sampel, $\hat{p}$)!
3. Sebutkan apakah penentuan angka $\hat{p}$ tersebut termasuk statistika deskriptif atau inferensial!

**Langkah Penyelesaian:**
1. **Populasi**: Seluruh orang dewasa di daerah tersebut.
   **Parameter**: Proporsi sebenarnya dari seluruh orang dewasa di daerah tersebut yang memiliki ponsel ($p$).
2. **Perhitungan Statistik Sampel ($\hat{p}$)**:
   $$\hat{p} = \frac{x}{n}$$
   Di mana $x = 1298$ dan $n = 1572$.
   $$\hat{p} = \frac{1298}{1572} \approx 0{,}8257 \text{ atau } 82{,}57\%$$
3. Menghitung angka $\hat{p} = 0{,}8257$ untuk merangkum data sampel adalah bagian dari **Statistika Deskriptif**. Apabila angka ini digunakan untuk menyatakan bahwa sekitar $82{,}57\%$ dari seluruh populasi memiliki ponsel, maka tindakan tersebut adalah **Statistika Inferensial**.

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulasi Penarikan Sampel (Populasi vs Sampel)
np.random.seed(42)

# 1. Buat Data Populasi (misal: usia kendaraan dengan distribusi tertentu)
population_data = np.random.gamma(shape=2, scale=3, size=100000)
pop_mean = np.mean(population_data)

# 2. Ambil Sampel Acak ukuran n = 200
sample_size = 200
sample_data = np.random.choice(population_data, size=sample_size, replace=False)
sample_mean = np.mean(sample_data)

print(f"Rata-rata Populasi (Mu)   : {pop_mean:.4f}")
print(f"Rata-rata Sampel (x-bar)  : {sample_mean:.4f}")

# Visualisasi
plt.figure(figsize=(10, 5))
plt.hist(population_data, bins=50, density=True, alpha=0.5, color='blue', label='Populasi')
plt.hist(sample_data, bins=30, density=True, alpha=0.7, color='orange', label=f'Sampel (n={sample_size})')
plt.axvline(pop_mean, color='blue', linestyle='dashed', linewidth=2, label=f'Mean Populasi (\u03bc={pop_mean:.2f})')
plt.axvline(sample_mean, color='red', linestyle='solid', linewidth=2, label=f'Mean Sampel (\u03bar{{x}}={sample_mean:.2f})')
plt.title("Visualisasi Perbandingan Distribusi Populasi dan Sampel")
plt.xlabel("Pengukuran")
plt.ylabel("Densitas")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

## 1.3. Presentation of Data
Penyajian data sampel dapat dilakukan dalam dua bentuk utama:
1. **Daftar Data (Data List)**: Pencatatan seluruh pengukuran secara eksplisit.
   Contoh: $\{18, 18, 19, 19, 19, 18, 22, 20, 18, 18, 17, 19, 18, 24, 18, 20, 18, 21, 20, 17, 19\}$.
2. **Tabel Frekuensi Data (Data Frequency Table)**: Penyajian data dengan mencantumkan setiap nilai unik $x$ pada baris/kolom pertama dan frekuensi kemunculannya $f$ pada baris/kolom kedua.

Rumus rata-rata sampel dari tabel frekuensi:
$$\bar{x} = \frac{\sum (x \cdot f)}{n}$$
Di mana $n = \sum f$ adalah total ukuran sampel.

## 1.4. Contoh Soal & Pembahasan
**Soal:**
Diberikan data usia dari 21 mahasiswa sebagai berikut:
`18, 18, 19, 19, 19, 18, 22, 20, 18, 18, 17, 19, 18, 24, 18, 20, 18, 21, 20, 17, 19`

1. Buatlah tabel frekuensi dari data tersebut!
2. Hitunglah rata-rata usia sampel ($\bar{x}$) menggunakan tabel frekuensi tersebut!

**Langkah Penyelesaian:**
1. **Membuat Tabel Frekuensi**:
   Hitung frekuensi $f$ untuk tiap nilai unik $x$:
   - $x = 17 \rightarrow f = 2$
   - $x = 18 \rightarrow f = 8$
   - $x = 19 \rightarrow f = 5$
   - $x = 20 \rightarrow f = 3$
   - $x = 21 \rightarrow f = 1$
   - $x = 22 \rightarrow f = 1$
   - $x = 24 \rightarrow f = 1$

   Total sampel: $n = \sum f = 2 + 8 + 5 + 3 + 1 + 1 + 1 = 21$.

2. **Menghitung Rata-rata Sampel ($\bar{x}$)**:
   $$\sum (x \cdot f) = (17 \times 2) + (18 \times 8) + (19 \times 5) + (20 \times 3) + (21 \times 1) + (22 \times 1) + (24 \times 1)$$
   $$\sum (x \cdot f) = 34 + 144 + 95 + 60 + 21 + 22 + 24 = 400$$

   $$\bar{x} = \frac{\sum (x \cdot f)}{n} = \frac{400}{21} \approx 19{,}0476$$
   Jadi, rata-rata usia mahasiswa dalam sampel tersebut adalah $19{,}05$ tahun.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Data usia mahasiswa
data = [18, 18, 19, 19, 19, 18, 22, 20, 18, 18, 17, 19, 18, 24, 18, 20, 18, 21, 20, 17, 19]

# Membuat Tabel Frekuensi menggunakan Pandas
df = pd.Series(data).value_counts().sort_index().reset_index()
df.columns = ['Usia (x)', 'Frekuensi (f)']
df['x * f'] = df['Usia (x)'] * df['Frekuensi (f)']

print("=== TABEL FREKUENSI ===")
print(df.to_string(index=False))

# Perhitungan Rata-rata
total_n = df['Frekuensi (f)'].sum()
sum_xf = df['x * f'].sum()
mean_val = sum_xf / total_n

print("\n=== PERHITUNGAN RATA-RATA ===")
print(f"Total Sampel (n) = {total_n}")
print(f"Sum(x * f)       = {sum_xf}")
print(f"Rata-rata (x̄)   = {mean_val:.4f}")

# Visualisasi Frekuensi menggunakan Diagram Batang
plt.figure(figsize=(8, 4))
plt.bar(df['Usia (x)'], df['Frekuensi (f)'], color='skyblue', edgecolor='black')
plt.xlabel('Usia (x)')
plt.ylabel('Frekuensi (f)')
plt.title('Penyajian Data Usia Mahasiswa (Diagram Batang)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
