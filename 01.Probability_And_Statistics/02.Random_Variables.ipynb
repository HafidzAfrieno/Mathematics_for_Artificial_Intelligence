## 1. Definisi Variabel Acak dan Fungsi Distribusi Kumulatif (CDF)

* **Variabel Acak**: Secara matematis, variabel acak $X$ adalah fungsi/pemetaan dari ruang sampel $\Omega$ ke himpunan bilangan real $\mathbb{R}$:
  $$X: \Omega \rightarrow \mathbb{R}$$
* **Fungsi Distribusi Kumulatif (CDF)**: Fungsi $F_X: \mathbb{R} \rightarrow [0, 1]$ yang mendefinisikan peluang bahwa variabel acak $X$ bernilai kurang dari atau sama dengan $x$:
  $$F_X(x) = \mathbb{P}(X \le x)$$
* **Sifat-sifat Utama CDF**:
  1. $F_X$ bersifat **tidak turun** (*non-decreasing*): Jika $x_1 < x_2$, maka $F_X(x_1) \le F_X(x_2)$.
  2. $F_X$ ter-**normalisasi**: $\lim_{x \to -\infty} F_X(x) = 0$ dan $\lim_{x \to \infty} F_X(x) = 1$.
  3. $F_X$ bersifat **kontinu dari kanan** (*right-continuous*): $F_X(x) = F_X(x^+)$ untuk semua $x$.

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

sample_space = ['TT', 'TH', 'HT', 'HH']
X_values = [0, 1, 1, 2] # Nilai X(omega) untuk setiap outcome

x_points = np.array([-1, 0, 1, 2, 3])
cdf_values = np.array([np.mean(np.array(X_values) <= x) for x in x_points])

print("--- 1. CDF Diskrit Pelemparan Koin 2 Kali ---")
for x, cdf in zip(x_points, cdf_values):
    print(f"F_X({x:d}) = P(X <= {x:d}) = {cdf:.2f}")
print("\n")
```

## 2. Variabel Acak Diskrit dan Kontinu

* **Variabel Acak Diskrit**: Mengambil nilai-nilai dalam himpunan terhitung $\{x_1, x_2, \dots\}$.
  * **Probability Mass Function (PMF)**: $f_X(x) = \mathbb{P}(X = x)$.
  * Hubungan dengan CDF: 
    $$F_X(x) = \sum_{x_i \le x} f_X(x_i)$$
* **Variabel Acak Kontinu**: Memiliki fungsi densitas peluang (*Probability Density Function* / PDF) $f_X(x) \ge 0$ sedemikian rupa sehingga $\int_{-\infty}^{\infty} f_X(x) dx = 1$.
  * Peluang pada rentang $[a, b]$:
    $$\mathbb{P}(a < X < b) = \int_{a}^{b} f_X(x) dx$$
  * Hubungan PDF dan CDF:
    $$F_X(x) = \int_{-\infty}^{x} f_X(t) dt \quad \text{dan} \quad f_X(x) = F_X'(x)$$
  * Catatan: Untuk variabel acak kontinu, peluang pada satu titik spesifik adalah nol, $\mathbb{P}(X = x) = 0$.

```python
a, b = -1.0, 1.0
prob_continuous = stats.norm.cdf(b, loc=0, scale=1) - stats.norm.cdf(a, loc=0, scale=1)

print("--- 2. Peluang Variabel Acak Kontinu ---")
print(f"P({a:.2f} < X < {b:.2f}) untuk Normal(0, 1) = {prob_continuous:.4f}\n")
```

## 3. Distribusi Diskrit dan Kontinu Penting

* **Distribusi Bernoulli**: $X \sim \text{Bernoulli}(p)$ dengan PMF:
  $$f_X(x) = p^x (1-p)^{1-x}, \quad x \in \{0, 1\}$$
* **Distribusi Binomial**: $X \sim \text{Binomial}(n, p)$ dengan PMF:
  $$f_X(x) = \binom{n}{x} p^x (1-p)^{n-x}, \quad x \in \{0, 1, \dots, n\}$$
* **Distribusi Poisson**: $X \sim \text{Poisson}(\lambda)$ dengan PMF:
  $$f_X(x) = \frac{e^{-\lambda} \lambda^x}{x!}, \quad x \in \{0, 1, 2, \dots\}$$
* **Distribusi Seragam Kontinu**: $X \sim \text{Uniform}(a, b)$ dengan PDF:
  $$f_X(x) = \frac{1}{b - a}, \quad x \in [a, b]$$
* **Distribusi Normal (Gaussian)**: $X \sim N(\mu, \sigma^2)$ dengan PDF:
  $$f_X(x) = \frac{1}{\sigma \sqrt{2\pi}} \exp \left( -\frac{1}{2\sigma^2} (x - \mu)^2 \right)$$

```python
# Binomial(n=10, p=0.5) dan Normal(mu=0, sigma=1)
n_bin, p_bin = 10, 0.5
x_bin = 5
pmf_bin = stats.binom.pmf(x_bin, n_bin, p_bin)

mu_norm, sigma_norm = 0.0, 1.0
x_norm = 1.96
pdf_norm = stats.norm.pdf(x_norm, mu_norm, sigma_norm)

print("--- 3. Evaluasi Distribusi Penting ---")
print(f"PMF Binomial(n=10, p=0.5) pada x={x_bin}: {pmf_bin:.4f}")
print(f"PDF Normal(mu=0, sigma=1) pada x={x_norm:.2f}: {pdf_norm:.4f}\n")
```

## 4. Distribusi Bivariat, Marginal, dan Kondisional

* **Bivariat / Joint Distribution**: Menggabungkan dua variabel acak $(X, Y)$.
  * Joint PDF/PMF: $f_{X,Y}(x, y)$.
  * Joint CDF: $F_{X,Y}(x, y) = \mathbb{P}(X \le x, Y \le y)$.
* **Distribusi Marginal**: Distribusi satu variabel dengan mengintegrasikan/menjumlahkan variabel lainnya.
  $$f_X(x) = \int_{-\infty}^{\infty} f_{X,Y}(x, y) dy \quad \text{atau} \quad f_X(x) = \sum_{y} f_{X,Y}(x, y)$$
* **Independensi Variabel Acak**: $X$ dan $Y$ independen ($X \perp\!\!\!\perp Y$) jika dan hanya jika:
  $$f_{X,Y}(x, y) = f_X(x) f_Y(y) \quad \forall x, y$$
* **Distribusi Kondisional**: Densitas/massa peluang $X$ ketika $Y = y$ diketahui:
  $$f_{X|Y}(x|y) = \frac{f_{X,Y}(x, y)}{f_Y(y)}, \quad \text{jika } f_Y(y) > 0$$

```python
# Distribusi bivariat diskrit: f(x, y) = (x + y) / 21 untuk x in {1, 2}, y in {1, 2, 3}
joint_pmf = np.zeros((2, 3))
for x_i in range(1, 3):
    for y_j in range(1, 4):
        joint_pmf[x_i - 1, y_j - 1] = (x_i + y_j) / 21.0

# Marginal X dan Marginal Y
marginal_X = np.sum(joint_pmf, axis=1)
marginal_Y = np.sum(joint_pmf, axis=0)

# Kondisional f_{X|Y}(x | Y=2)
conditional_X_given_Y2 = joint_pmf[:, 1] / marginal_Y[1]

print("--- 4. Distribusi Bivariat & Marginal ---")
print(f"Joint PMF Matrix (Baris: X=1,2 | Kolom: Y=1,2,3):\n{joint_pmf}")
print(f"Marginal PMF X (x=1, 2): {marginal_X[0]:.3f}, {marginal_X[1]:.3f}")
print(f"Conditional PMF X | Y=2 (x=1, 2): {conditional_X_given_Y2[0]:.3f}, {conditional_X_given_Y2[1]:.3f}\n")
```

## 5. Transformasi Variabel Acak

Jika $Y = r(X)$ adalah fungsi dari variabel acak $X$:
1. **Langkah 1**: Tentukan daerah himpunan $A_y = \{x : r(x) \le y\}$.
2. **Langkah 2**: Cari CDF dari $Y$:
   $$F_Y(y) = \mathbb{P}(Y \le y) = \mathbb{P}(r(X) \le y) = \int_{A_y} f_X(x) dx$$
3. **Langkah 3**: Turunkan $F_Y(y)$ untuk mendapatkan PDF $f_Y(y) = F_Y'(y)$.
* Jika $r$ monoton murni dan memiliki invers $s = r^{-1}$:
  $$f_Y(y) = f_X(s(y)) \left| \frac{ds(y)}{dy} \right|$$

```python
# Menggunakan Probability Integral Transform: U ~ Uniform(0, 1) -> X = -beta * ln(1 - U) ~ Exp(beta)
np.random.seed(42)
num_samples = 10000
beta = 2.0

U = np.random.uniform(0, 1, num_samples)
X_simulated = -beta * np.log(1 - U)

mean_simulated = np.mean(X_simulated)
mean_theoretical = beta

print("--- 5. Simulasi Transformasi Variabel Acak (Exponential) ---")
print(f"Rata-rata hasil simulasi X: {mean_simulated:.3f}")
print(f"Rata-rata teoritis Eksponesial(beta={beta:.1f}): {mean_theoretical:.3f}\n")
```
