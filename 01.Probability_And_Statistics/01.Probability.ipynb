# **Bab 1: Dasar-Dasar Probabilitas**

## 1. Ruang Sampel dan Kejadian
* **Ruang Sampel ($\Omega$)**: Himpunan semua kemungkinan hasil dari suatu eksperimen.
* **Kejadian ($A$)**: Subhimpunan dari ruang sampel ($A \subseteq \Omega$).
* **Operasi Dasar**:
  * Gabungan (Union): $A \cup B$ (A atau B terjadi)
  * Irisan (Intersection): $A \cap B$ (A dan B terjadi bersamaan)
  * Komplemen ($A^c$): Kejadian di mana $A$ tidak terjadi.
  * Saling Lepas (Mutually Exclusive): $A \cap B = \emptyset$

```python
import math
Omega = {1, 2, 3, 4, 5, 6} # Eksperimen lempar 1 dadu
A = {2, 4, 6}  # Kejadian genap
B = {4, 5, 6}  # Kejadian > 3

print("--- 1. OPERASI HIMPUNAN ---")
print(f"Ruang Sampel Omega : {Omega}")
print(f"A ∪ B (Gabungan)   : {A | B}")
print(f"A ∩ B (Irisan)     : {A & B}")
print(f"A^c (Komplemen A)  : {Omega - A}\n")
```

## 2. Aksioma Peluang (Kolmogorov)
Peluang $\mathbb{P}$ memetakan kejadian ke bilangan real dengan 3 syarat:
1. **Aksioma 1**: $\mathbb{P}(A) \ge 0$
2. **Aksioma 2**: $\mathbb{P}(\Omega) = 1$
3. **Aksioma 3**: Jika kejadian-kejadian saling lepas, maka peluang gabungannya adalah jumlah peluang masing-masing: $$\mathbb{P}\left(\bigcup_{i=1}^{\infty} A_i\right) = \sum_{i=1}^{\infty} \mathbb{P}(A_i)$$

```python
def prob(event, sample_space):
    return len(event) / len(sample_space)

P_A = prob(A, Omega)
P_B = prob(B, Omega)
P_A_irisan_B = prob(A & B, Omega)

# Aturan Penjumlahan: P(A U B) = P(A) + P(B) - P(A ∩ B)
P_A_gabungan_B = P_A + P_B - P_A_irisan_B

print("--- 2. PELUANG & ATURAN PENJUMLAHAN ---")
print(f"P(A) = {P_A:.2f}, P(B) = {P_B:.2f}")
print(f"P(A ∪ B) = {P_A_gabungan_B:.2f}\n")
```

## 3. Kombinatorika & Ruang Sampel Hingga
Jika setiap elemen dalam $\Omega$ memiliki peluang muncul yang sama (Uniform):
$$\mathbb{P}(A) = \frac{|A|}{|\Omega|}$$
* **Faktorial ($n!$)**: Susunan $n$ objek yang memperhatikan urutan.
* **Kombinasi ($\binom{n}{k}$)**: Memilih $k$ objek dari $n$ objek tanpa memperhatikan urutan: $$\binom{n}{k} = \frac{n!}{k!(n-k)!}$$

```python
# Contoh: Peluang memilih 2 bola merah dari 5 bola merah
n_total_bola = 10
n_bola_merah = 5
k_pilih = 2

# Kombinasi (n choose k)
total_cara = math.comb(n_total_bola, k_pilih)
cara_merah = math.comb(n_bola_merah, k_pilih)
P_merah = cara_merah / total_cara

print("--- 3. KOMBINATORIKA ---")
print(f"Banyak cara pilih 2 bola dari 10 : {total_cara}")
print(f"Peluang terpilih 2 bola merah    : {P_merah:.3f}\n")
```

## 4. Independensi & Peluang Bersyarat
* **Kejadian Independen**: Kejadian $A$ tidak mempengaruhi peluang terjadinya $B$. $$\mathbb{P}(A \cap B) = \mathbb{P}(A)\mathbb{P}(B)$$
* **Peluang Bersyarat**: Peluang $A$ terjadi jika diketahui $B$ telah terjadi ($\mathbb{P}(B) > 0$). $$\mathbb{P}(A|B) = \frac{\mathbb{P}(A \cap B)}{\mathbb{P}(B)}$$

```python
# Peluang bersyarat P(A|B) = P(A ∩ B) / P(B)
P_A_given_B = P_A_irisan_B / P_B

# Cek independensi P(A ∩ B) == P(A) * P(B)
is_independent = math.isclose(P_A_irisan_B, P_A * P_B)

print("--- 4. BERSYARAT & INDEPENDENSI ---")
print(f"P(A|B) (Peluang genap jika dadu > 3): {P_A_given_B:.2f}")
print(f"Apakah A dan B independen? {is_independent}\n")
```

## 5. Hukum Peluang Total & Teorema Bayes
Jika $A_1, A_2, \dots, A_k$ mempartisi ruang sampel $\Omega$:
* **Hukum Peluang Total**: $$\mathbb{P}(B) = \sum_{i=1}^{k} \mathbb{P}(B|A_i)\mathbb{P}(A_i)$$
* **Teorema Bayes**: Digunakan untuk mencari peluang *prior* setelah kejadian *posterior* diketahui: $$\mathbb{P}(A_i|B) = \frac{\mathbb{P}(B|A_i)\mathbb{P}(A_i)}{\mathbb{P}(B)}$$

```python
P_Hujan = 0.3          # P(A) Prior
P_Cerah = 0.7          # P(A^c)
P_Macet_if_Hujan = 0.8 # P(B|A)
P_Macet_if_Cerah = 0.2 # P(B|A^c)

# Peluang Total P(Macet)
P_Macet = (P_Macet_if_Hujan * P_Hujan) + (P_Macet_if_Cerah * P_Cerah)

# Teorema Bayes: P(Hujan|Macet)
P_Hujan_given_Macet = (P_Macet_if_Hujan * P_Hujan) / P_Macet

print("--- 5. TEOREMA BAYES ---")
print(f"Peluang Total Macet       : {P_Macet:.2f}")
print(f"Peluang Hujan JIKA Macet  : {P_Hujan_given_Macet:.2f}")
```

## 1. Latihan Soal Kombinasi dan Permutasi

1. Dalam sebuah perlombaan lari yang diikuti oleh 8 orang peserta, akan dipilih juara I, juara II, dan juara III. Berapa banyak susunan pemenang yang mungkin terjadi?

```python
kombinasi_juara = math.perm(8,3)
print(f'Banyak susunan adalah {kombinasi_juara}')
```

2. Sebuah organisasi memiliki 10 orang anggota. Jika pengurus ingin membentuk panitia kecil yang terdiri dari 4 orang untuk mengatur suatu acara, berapa banyak cara memilih panitia tersebut?

```python
kombinasi_panitia = math.comb(10,4)
print(f'banyak cara memilih panitia adalah {kombinasi_panitia}')
```

3. Seseorang ingin membuat kode PIN ATM yang terdiri dari 4 angka berbeda dari digit 1,2,3,4,5,6,dan 7. Berapa banyak variasi kode PIN yang dapat dibuat?

```python
kombinasi_PIN = math.perm(7,4)
print(f'Kombinasi PIN adalah {kombinasi_PIN}')
```

4. Di sebuah klub bulu tangkis terdapat 6 pemain putra dan 5 pemain putri. Pelatih ingin membentuk pasangan ganda campuran (1 putra dan 1 putri) untuk mengikuti sebuah turnamen. Berapa banyak pasangan ganda campuran berbeda yang bisa dibentuk?

```python
comb_putra = math.comb(6,1)
comb_putri = math.comb(5,1)
banyak_pasangan = comb_putra * comb_putri
print(f'banyak pasangan ganda yang bisa terbentuk adalah {banyak_pasangan}')
```

5. Sebuah kelompok yang terdiri dari 5 orang murid ingin berfoto bersama secara berdampingan dalam satu baris. Berapa banyak posisi berdiri berbeda yang dapat dibentuk saat foto bersama?

```python
banyak_posisi = math.factorial(5)
print(f'banyak posisi berdiri adalah {banyak_posisi}')
```

6. Berapa banyak cara membentuk sebuahpanitia yang beranggotakan 5 orang yangdipilih dari 7 orang pria dan 5 orang wanita,jika di dalam panitia tersebut paling sedikitberanggotakan 2 orang wanita?

```python
total_comb = 0
for i in range (2,6):
    comb_pria = math.comb(5,i)
    comb_wanita = math.comb(7,5-i)
    total_comb += (comb_pria*comb_wanita)
print(f'total kombinasi adalah {total_comb}')
```
