# Chapter 1: Probability

## 1. Ruang Sampel dan Kejadian

### 1. Ruang Sampel ($\Omega$): Himpunan seluruh hasil  Elemen di dalamnya disebut $\omega$

```python
Omega = {1, 2, 3, 4, 5, 6}
print(f"Ruang Sampel (Omega): {Omega}")
```

### 2. Kejadian (Event, $A$): Subhimpunan dari ruang sampel $\Omega$ ($A \subseteq \Omega$)

```python
A = {2, 4, 6}  # Kejadian muncul angka genap
B = {4, 5, 6}  # Kejadian muncul angka lebih besar dari 3
C = {1, 3, 5}  # Kejadian muncul angka ganjil
print(f"Kejadian A (Genap) : {A}")
print(f"Kejadian B (> 3)   : {B}")
print(f"Kejadian C (Ganjil): {C}\n")
```

### 3. Operasi Himpunan Utama

 1. Komplemen ($A^c$):

```python
A_komplemen = Omega - A
print(f"Komplemen A (A^c)  : {A_komplemen}")
```

2. Gabungan ($A \cup B$):

```python
A_gabungan_B = A | B  # Bisa juga A.union(B)
print(f"Gabungan (A ∪ B)   : {A_gabungan_B}")
```

3. Irisan ($A \cap B$ atau $AB$):

```python
A_irisan_B = A & B  
print(f"Irisan (A ∩ B)     : {A_irisan_B}\n")
```

4. Saling Lepas (Disjoint) Kejadian $A$ dan $B$ tidak memiliki elemen bersama:

```python
is_saling_lepas = A.isdisjoint(C) 
print(f"Apakah A dan C saling lepas? {is_saling_lepas} (Irisan: {A & C})")
```

5. Partisi $\bigcup_{i=1}^{\infty} A_i = \Omega$ (Urutan himpunan-himpunan saling lepas):

```python
is_partisi = A.isdisjoint(C) and (A | C == Omega)
print(f"Apakah A dan C membentuk partisi dari Omega? {is_partisi}")
```

6. Fungsi Indikator ($I_A$):

```python
def indikator(omega, himpunan):
    return 1 if omega in himpunan else 0

hasil_lemparan = 4
hasil_lemparan_2 = 5

print("Fungsi Indikator untuk Kejadian A (Angka Genap):")
print(f"I_A({hasil_lemparan}) = {indikator(hasil_lemparan, A)}")
print(f"I_A({hasil_lemparan_2}) = {indikator(hasil_lemparan_2, A)}")
```

## 2. Aksioma Peluang dan Aturan-Aturannya

```python
Omega = {1, 2, 3, 4, 5, 6} 
himpunan_kosong = set()

def P(kejadian):
    if not kejadian: 
        return 0.0
    return len(kejadian) / len(Omega)

A = {1, 2, 3}
B = {4, 5}  
C = {3, 4, 5}
```

### 1. Aksioma Peluang

 1. Aksioma 1: $\mathbb{P}(A) \ge 0$ untuk setiap kejadian $A$

```python
print(f"Aksioma 1: P(A) = {P(A):.2f} (Memenuhi P(A) >= 0: {P(A) >= 0})")
```

2. Aksioma 2: $\mathbb{P}(\Omega) = 1$

```python
print(f"Aksioma 2: P(Omega) = {P(Omega):.2f} (Memenuhi P(Omega) == 1: {P(Omega) == 1})")
```

3. Aksioma 3: Jika $A_1, A_2, \dots$ saling lepas (disjoint), maka $\mathbb{P}\left(\bigcup_{i=1}^{\infty} A_i\right) = \sum_{i=1}^{\infty} \mathbb{P}(A_i)$

```python
is_disjoint = A.isdisjoint(B)
P_A_gabung_B = P(A | B)
P_A_tambah_P_B = P(A) + P(B)
print(f"Aksioma 3: A dan B saling lepas? {is_disjoint}")
print(f"           P(A ∪ B)     = {P_A_gabung_B:.2f}")
print(f"           P(A) + P(B)  = {P_A_tambah_P_B:.2f}\n")
```
