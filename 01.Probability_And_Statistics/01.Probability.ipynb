# **Bab 1: Dasar-Dasar Probabilitas**

## 1. Ruang Sampel dan Kejadian
* **Ruang Sampel ($\Omega$)**: Himpunan semua kemungkinan hasil dari suatu eksperimen.
* **Kejadian ($A$)**: Subhimpunan dari ruang sampel ($A \subseteq \Omega$).
* **Operasi Dasar**:
  * Gabungan (Union): $A \cup B$ (A atau B terjadi)
  * Irisan (Intersection): $A \cap B$ (A dan B terjadi bersamaan)
  * Komplemen ($A^c$): Kejadian di mana $A$ tidak terjadi.
  * Saling Lepas (Mutually Exclusive): $A \cap B = \emptyset$

```python
import math
Omega = {1, 2, 3, 4, 5, 6} # Eksperimen lempar 1 dadu
A = {2, 4, 6}  # Kejadian genap
B = {4, 5, 6}  # Kejadian > 3

print("--- 1. OPERASI HIMPUNAN ---")
print(f"Ruang Sampel Omega : {Omega}")
print(f"A ∪ B (Gabungan)   : {A | B}")
print(f"A ∩ B (Irisan)     : {A & B}")
print(f"A^c (Komplemen A)  : {Omega - A}\n")
```

## 2. Aksioma Peluang (Kolmogorov)
Peluang $\mathbb{P}$ memetakan kejadian ke bilangan real dengan 3 syarat:
1. **Aksioma 1**: $\mathbb{P}(A) \ge 0$
2. **Aksioma 2**: $\mathbb{P}(\Omega) = 1$
3. **Aksioma 3**: Jika kejadian-kejadian saling lepas, maka peluang gabungannya adalah jumlah peluang masing-masing: $$\mathbb{P}\left(\bigcup_{i=1}^{\infty} A_i\right) = \sum_{i=1}^{\infty} \mathbb{P}(A_i)$$

```python
def prob(event, sample_space):
    return len(event) / len(sample_space)

P_A = prob(A, Omega)
P_B = prob(B, Omega)
P_A_irisan_B = prob(A & B, Omega)

# Aturan Penjumlahan: P(A U B) = P(A) + P(B) - P(A ∩ B)
P_A_gabungan_B = P_A + P_B - P_A_irisan_B

print("--- 2. PELUANG & ATURAN PENJUMLAHAN ---")
print(f"P(A) = {P_A:.2f}, P(B) = {P_B:.2f}")
print(f"P(A ∪ B) = {P_A_gabungan_B:.2f}\n")
```

## 3. Kombinatorika & Ruang Sampel Hingga
Jika setiap elemen dalam $\Omega$ memiliki peluang muncul yang sama (Uniform):
$$\mathbb{P}(A) = \frac{|A|}{|\Omega|}$$
* **Faktorial ($n!$)**: Susunan $n$ objek yang memperhatikan urutan.
* **Kombinasi ($\binom{n}{k}$)**: Memilih $k$ objek dari $n$ objek tanpa memperhatikan urutan: $$\binom{n}{k} = \frac{n!}{k!(n-k)!}$$

```python
# Contoh: Peluang memilih 2 bola merah dari 5 bola merah
n_total_bola = 10
n_bola_merah = 5
k_pilih = 2

# Kombinasi (n choose k)
total_cara = math.comb(n_total_bola, k_pilih)
cara_merah = math.comb(n_bola_merah, k_pilih)
P_merah = cara_merah / total_cara

print("--- 3. KOMBINATORIKA ---")
print(f"Banyak cara pilih 2 bola dari 10 : {total_cara}")
print(f"Peluang terpilih 2 bola merah    : {P_merah:.3f}\n")
```

## 4. Independensi & Peluang Bersyarat
* **Kejadian Independen**: Kejadian $A$ tidak mempengaruhi peluang terjadinya $B$. $$\mathbb{P}(A \cap B) = \mathbb{P}(A)\mathbb{P}(B)$$
* **Peluang Bersyarat**: Peluang $A$ terjadi jika diketahui $B$ telah terjadi ($\mathbb{P}(B) > 0$). $$\mathbb{P}(A|B) = \frac{\mathbb{P}(A \cap B)}{\mathbb{P}(B)}$$

```python
# Peluang bersyarat P(A|B) = P(A ∩ B) / P(B)
P_A_given_B = P_A_irisan_B / P_B

# Cek independensi P(A ∩ B) == P(A) * P(B)
is_independent = math.isclose(P_A_irisan_B, P_A * P_B)

print("--- 4. BERSYARAT & INDEPENDENSI ---")
print(f"P(A|B) (Peluang genap jika dadu > 3): {P_A_given_B:.2f}")
print(f"Apakah A dan B independen? {is_independent}\n")
```

## 5. Hukum Peluang Total & Teorema Bayes
Jika $A_1, A_2, \dots, A_k$ mempartisi ruang sampel $\Omega$:
* **Hukum Peluang Total**: $$\mathbb{P}(B) = \sum_{i=1}^{k} \mathbb{P}(B|A_i)\mathbb{P}(A_i)$$
* **Teorema Bayes**: Digunakan untuk mencari peluang *prior* setelah kejadian *posterior* diketahui: $$\mathbb{P}(A_i|B) = \frac{\mathbb{P}(B|A_i)\mathbb{P}(A_i)}{\mathbb{P}(B)}$$

```python
P_Hujan = 0.3          # P(A) Prior
P_Cerah = 0.7          # P(A^c)
P_Macet_if_Hujan = 0.8 # P(B|A)
P_Macet_if_Cerah = 0.2 # P(B|A^c)

# Peluang Total P(Macet)
P_Macet = (P_Macet_if_Hujan * P_Hujan) + (P_Macet_if_Cerah * P_Cerah)

# Teorema Bayes: P(Hujan|Macet)
P_Hujan_given_Macet = (P_Macet_if_Hujan * P_Hujan) / P_Macet

print("--- 5. TEOREMA BAYES ---")
print(f"Peluang Total Macet       : {P_Macet:.2f}")
print(f"Peluang Hujan JIKA Macet  : {P_Hujan_given_Macet:.2f}")
```

1. Sebuah pabrik menggunakan dua unit mesin produksi, yaitu Mesin X dan Mesin Y. Probabilitas Mesin X beroperasi dengan normal adalah $0{,}95$, 
sedangkan probabilitas Mesin Y beroperasi dengan normal adalah $0{,}90$. Apabila kinerja kedua mesin tidak saling mempengaruhi:

```python
# a.Berapa probabilitas kedua mesin tersebut dapat beroperasi secara normal? 
p_mesin_x = 0.95
p_mesin_y = 0.90
probailitas_mesin_Xy = p_mesin_x * p_mesin_y
print(f'probabilitas kedua mesin tersebut dapat beroperasi secara normal adalah {probailitas_mesin_Xy}')
```

```python
# b.Berapa probabilitas hanya satu dari kedua mesin tersebut yang beroperasi secara normal?
p_xc = 1 - p_mesin_x
p_yc = 1 - p_mesin_y
x_n_yc = p_mesin_x * p_yc
y_n_xc = p_mesin_y * p_xc
total_proba = y_n_xc + x_n_yc
print(f'probabilitas hanya satu dari kedua mesin tersebut yang beroperasi secara normal adalah {total_proba}')
```

2. Dua buah sistem radar digunakan untuk mendeteksi ancaman penerbangan. Probabilitas radar A mampu mendeteksi sasaran tepat waktu adalah $0{,}80$, sedangkan probabilitas radar B adalah $0{,}85$. Jika keberhasilan tiap radar independen satu sama lain:

```python
# a. Berapa probabilitas kedua radar gagal mendeteksi sasaran tersebut?
radar_a = 0.80
radar_b = 0.85
radar_a_rusak = 1 - radar_a
radar_b_rusak = 1 - radar_b
radar_ab_rusak = radar_a_rusak * radar_b_rusak
print(f'probabilitas kedua radar gagal mendeteksi sasaran adalah {radar_ab_rusak}')
```

```python
# b. Berapa probabilitas hanya radar A yang berhasil mendeteksi sasaran?
hanya_radar_a_rusak = radar_a * radar_b_rusak
print(f' probabilitas hanya radar A yang berhasil mendeteksi sasaran adalah {round(hanya_radar_a_rusak,3)}')
```

3. Seorang kandidat mengikuti ujian lisensi profesi yang terdiri dari tiga modul independent: Modul I, Modul II, dan Modul III. Probabilitas kelulusannya pada masing-masing modul berturut-turut adalah $0{,}70$, $0{,}80$, dan $0{,}50$. Hitunglah probabilitas kandidat tersebut:

```python
# a. Berapa probabilitas kedua radar gagal mendeteksi sasaran tersebut?
modul_a = 0.7
modul_b = 0.8
modul_c = 0.5
lulus_modul_abc = modul_a * modul_b * modul_c
print(f'Peluang Lulus dalam ketiga modul tersebut adalah {lulus_modul_abc}')
```

```python
# b. Tidak lulus dalam semua modul tersebut.
modul_ac = 1 - modul_a
modul_bc = 1 - modul_b
modul_cc = 1 - modul_c
tidak_lulus_modul_abc = modul_ac * modul_bc * modul_cc
print(f'Peluang Tidak lulus dalam semua modul tersebut adalah {tidak_lulus_modul_abc}')
```

```python
# c. Lulus tepat pada dua modul saja
lulus_modul_ab = modul_a * modul_b * modul_cc
lulus_modul_ac = modul_a * modul_c * modul_bc
lulus_modul_bc = modul_b * modul_c * modul_ac
total_lulus_2modul = lulus_modul_ab + lulus_modul_ac + lulus_modul_bc
print(f'Peluang Lulus tepat pada dua modul saja adalah {total_lulus_2modul}')
```

4. Sebuah pasangan merencanakan untuk memiliki 4 orang anak. Jika asumsi peluang melahirkan anak laki-laki dan perempuan adalah sama:

```python
# a. Berapa probabilitas mereka memiliki anak perempuan semua? 
jumlah_variasi_anak = 2 ** 4
prm_semua = math.comb(4,4)
peluang_prm_semua = prm_semua / jumlah_variasi_anak
print(f'probabilitas mereka memiliki anak perempuan semua adalah {peluang_prm_semua}')
```

```python
# b. Berapa probabilitas mereka memiliki tepat 2 anak laki-laki dan 2 anak perempuan?
kombinasi_2laki_2perempuan = math.comb(4,2)
peluang_laki_dan_prempuan = kombinasi_2laki_2perempuan / jumlah_variasi_anak
print(f'probabilitas mereka memiliki tepat 2 anak laki-laki dan 2 anak perempuan adalah {peluang_laki_dan_prempuan}')
```

```python
# c. Berapa probabilitas mereka memiliki paling sedikit satu anak laki-laki?
minimal_1_laki = 1 - peluang_prm_semua
print(f'probabilitas mereka memiliki paling sedikit satu anak laki-lak adalah {minimal_1_laki}')
```

5. Probabilitas seorang analis data melakukan pengolahan data dengan akurat adalah 0,85. Apabila analis tersebut membuat kesalahan pengolahan data, probabilitas klien akan mengajukan komplain adalah 0,75. Berapa probabilitas bahwa analis tersebut membuat kesalahan pengolahan data dan klien mengajukan komplain?

```python
akurasi_olah = 0.85
kesalahan_olah = 1 - akurasi_olah
klien_komplain = 0.75
peluang_kesalahanolah_dan_komplain = kesalahan_olah * klien_komplain
print(f'Pelunag Kesdalahan olah data dan klien mengajukan komplain adalah {peluang_kesalahanolah_dan_komplain}')
```

6. Suatu perusahaan transportasi memiliki armada truk dari tiga karoseri berbeda: 40%Armada A, 35%Armada B, dan 25%Armada C. Berdasarkan catatan histori operasional, tingkat kerusakan mesin pada Armada A adalah 2%, Armada B sebesar 5%, dan Armada C sebesar 4%.

```python
# a. Hitunglah total probabilitas sebuah truk yang dipilih secara acak mengalami kerusakan mesin.
armada_a = 0.4
armada_b = 0.35
armada_c = 0.25

rusak_a = 0.02
rusak_b = 0.05
rusak_c = 0.04

total_kerusakan_armada = (armada_a * rusak_a) + (armada_b * rusak_b) + (armada_c * rusak_c)
print(f'otal probabilitas sebuah truk yang dipilih secara acak mengalami kerusakan mesin adalah {total_kerusakan_armada}')
```

```python
# b. Jika sebuah truk yang dipilih secara acak ternyata mengalami kerusakan mesin, berapa probabilitas truk tersebut berasal dari Armada B?
armada_b_rusak =  (armada_b * rusak_b) / total_kerusakan_armada
print(f'Peluang Kerusakan dari armada B adalah {armada_b_rusak}')
```

7. Sebuah gudang menerima pasokan komponen elektronik dari tiga vendor berbeda: Vendor I menyuplai 50%, Vendor II menyuplai 30%, dan Vendor III menyuplai 20%. Persentase komponen cacat yang dihasilkan oleh Vendor I, II, dan III masing-masing adalah 3%, 5%, dan 8%.

```python
# a. Berapa probabilitas satu komponen yang diambil secara acak dari gudang tersebut cacat? 
vendor_sumpllier = [0.5,0.3,0.2]
rusak_vendor     = [0.03,0.05,0.08]
total_kerusakan_vendor = 0

for i in range(0,3):
    total_kerusakan_vendor += (vendor_sumpllier[i] * rusak_vendor[i])
print(f' probabilitas satu komponen cacat di gudang adalah: {total_kerusakan_vendor}')
```

```python
# b. Jika komponen yang diambil secara acak ternyata cacat, berapa probabilitas komponen tersebut dikirim oleh Vendor III?
komponen_cacat_vendor_b = (vendor_sumpllier[2] * rusak_vendor[2]) / total_kerusakan_vendor
print(f'Pelunang komponen cacat dari supplier B adalah {komponen_cacat_vendor_b}')
```
