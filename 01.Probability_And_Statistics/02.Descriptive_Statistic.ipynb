## 2.1. Visualisasi Data (Data Displays)

Dalam statistik deskriptif, menyajikan data secara visual sangat penting untuk memahami pola dan distribusi data. Tiga diagram populer yang sering digunakan adalah:

1. **Stem and Leaf Diagram (Diagram Batang dan Daun)**:
   Membagi setiap nilai data menjadi "batang" (puluhan/ratusan) dan "daun" (satuan). Berguna untuk data berukuran kecil hingga sedang karena mempertahankan nilai asli data.

2. **Frequency Histogram (Histogram Frekuensi)**:
   Diagram batang tegak lurus di mana sumbu horizontal ($x$) menyatakan kelas interval data dan sumbu vertikal ($y$) menyatakan frekuensi ($f$) atau jumlah kemunculan data dalam interval tersebut.

3. **Relative Frequency Histogram (Histogram Frekuensi Relatif)**:
   Serupa dengan histogram frekuensi, namun sumbu vertikal ($y$) menyatakan frekuensi relatif ($\frac{f}{n}$ atau persentase). Luas total seluruh batang pada histogram frekuensi relatif bernilai $1$ (atau $100\%$). Jika ukuran sampel $n \to \infty$, bentuk histogram frekuensi relatif akan mendekati kurva mulus (*density curve*).

---

## 2.2. Ukuran Pemusatan Data (Measures of Central Location)

Ukuran pemusatan data digunakan untuk menjawab pertanyaan: *"Di manakah titik pusat dari kumpulan data?"*

### 1. Mean (Rata-rata Hitung)
Rata-rata hitung diperoleh dengan menjumlahkan seluruh nilai data lalu membaginya dengan jumlah total observasi.

* **Sampel Mean ($\bar{x}$)**:
  $$\bar{x} = \frac{\sum x}{n}$$
  di mana $n$ adalah ukuran sampel dan $\sum x$ adalah jumlah seluruh nilai data sampel.

* **Populasi Mean ($\mu$)**:
  $$\mu = \frac{\sum x}{N}$$
  di mana $N$ adalah ukuran populasi.

* **Untuk Data Berfrekuensi**:
  $$\bar{x} = \frac{\sum (x \cdot f)}{n}$$

### 2. Median ($\tilde{x}$)
Nilai tengah dari sekumpulan data yang telah diurutkan dari nilai terkecil hingga terbesar. Median membagi data menjadi dua bagian sama besar ($50\%$ data di sebelah kiri dan $50\%$ di sebelah kanan).

* Jika jumlah data $n$ **ganjil**, median adalah data tepat di posisi tengah.
* Jika jumlah data $n$ **genap**, median adalah rata-rata dari dua data di posisi tengah:
  $$\tilde{x} = \frac{x_{\left(\frac{n}{2}\right)} + x_{\left(\frac{n}{2} + 1\right)}}{2}$$

* **Catatan Skewness (Kemiringan)**:
  * **Simetris**: Mean $\approx$ Median ($\bar{x} = \tilde{x}$)
  * **Skewed Right (Miring ke Kanan)**: Mean $>$ Median ($\bar{x} > \tilde{x}$) — ditarik oleh nilai ekstrem tinggi (*outliers*).
  * **Skewed Left (Miring ke Kiri)**: Mean $<$ Median ($\bar{x} < \tilde{x}$) — ditarik oleh nilai ekstrem rendah.

### 3. Mode (Modus)
Nilai data yang paling sering muncul (memiliki frekuensi terbesar). Kumpulan data dapat bersifat *unimodal* (satu modus), *bimodal* (dua modus), *multimodal*, atau tidak memiliki modus.

```python
# SEL CODE 1: Visualisasi Data & Ukuran Pemusatan Data
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Data contoh: Nilai ujian 30 mahasiswa
data_ujian = np.array([
    86, 80, 25, 77, 73, 76, 100, 90, 69, 93,
    90, 83, 70, 73, 73, 70, 90, 83, 71, 95,
    40, 58, 68, 69, 100, 78, 87, 97, 92, 74
])

# 1. Perhitungan Ukuran Pemusatan
mean_val = np.mean(data_ujian)
median_val = np.median(data_ujian)
mode_val = stats.mode(data_ujian, keepdims=True).mode[0]

print(f"--- Ukuran Pemusatan Data ---")
print(f"Mean (Rata-rata) : {mean_val:.2f}")
print(f"Median           : {median_val:.2f}")
print(f"Modus            : {mode_val}")

# 2. Visualisasi Histogram Frekuensi Relatif
plt.figure(figsize=(10, 5))
count, bins, patches = plt.hist(data_ujian, bins=8, edgecolor='black', alpha=0.7, density=True, color='skyblue')

# Menambahkan garis Vertikal untuk Mean dan Median
plt.axvline(mean_val, color='red', linestyle='dashed', linewidth=2, label=f'Mean ({mean_val:.2f})')
plt.axvline(median_val, color='green', linestyle='solid', linewidth=2, label=f'Median ({median_val:.2f})')

plt.title('Relative Frequency Histogram Ujian Matematika', fontsize=14)
plt.xlabel('Nilai Ujian', fontsize=12)
plt.ylabel('Frekuensi Relatif', fontsize=12)
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.show()
```

## 2.3. Ukuran Penyebaran Data (Measures of Variability)

Ukuran penyebaran digunakan untuk mengukur seberapa jauh nilai-nilai data menyebar dari pusatnya.

### 1. Range (Jangkauan)
Selisih antara nilai terbesar dan nilai terkecil dalam kumpulan data.
$$R = x_{\text{max}} - x_{\text{min}}$$

### 2. Varians (Variance)
Mengukur rata-rata penyimpangan kuadrat tiap data terhadap rata-ratanya.

* **Sampel Variance ($s^2$)**:
  $$s^2 = \frac{\sum (x - \bar{x})^2}{n - 1} = \frac{\sum x^2 - \frac{1}{n}\left(\sum x\right)^2}{n - 1}$$
  *(Pembagi $n-1$ digunakan agar varians sampel menjadi estimator tidak bias bagi populasi).*

* **Populasi Variance ($\sigma^2$)**:
  $$\sigma^2 = \frac{\sum (x - \mu)^2}{N}$$

### 3. Standar Deviasi (Simpangan Baku)
Akar kuadrat positif dari varians. Memiliki satuan yang sama dengan data aslinya.

* **Sampel Standard Deviation ($s$)**:
  $$s = \sqrt{s^2} = \sqrt{\frac{\sum (x - \bar{x})^2}{n - 1}}$$

* **Populasi Standard Deviation ($\sigma$)**:
  $$\sigma = \sqrt{\sigma^2} = \sqrt{\frac{\sum (x - \mu)^2}{N}}$$

---

## 2.4. Posisi Relatif Data (Relative Position of Data)

### 1. Percentile Rank & Quartiles
* **Percentile Rank**: Persentil ke-$P$ ($P^{\text{th}}$ percentile) adalah nilai $x$ di mana $P\%$ dari data bernilai kurang dari atau sama dengan $x$.
* **Quartiles (Kuartil)**: Membagi data yang terurut menjadi 4 bagian sama besar:
  * $Q_1$ (Kuartil Pertama / Bawah): Median dari paruh bawah data ($25\%$ data berada di bawah $Q_1$).
  * $Q_2$ (Kuartil Kedua / Tengah): Median dari seluruh data ($50\%$ data berada di bawah $Q_2$).
  * $Q_3$ (Kuartil Ketiga / Atas): Median dari paruh atas data ($75\%$ data berada di bawah $Q_3$).

* **Interquartile Range ($IQR$)**:
  $$\text{IQR} = Q_3 - Q_1$$

* **Five-Number Summary**:
  $$\{x_{\text{min}}, Q_1, Q_2, Q_3, x_{\text{max}}\}$$
  Ringkasan lima angka ini digunakan untuk membuat **Box Plot (Box-and-Whisker Plot)**.

### 2. z-Score (Skor Baku)
Mengukur berapa kali standar deviasi suatu nilai data $x$ berada di atas atau di bawah nilai rata-ratanya:
$$z = \frac{x - \mu}{\sigma} \quad \text{atau} \quad z = \frac{x - \bar{x}}{s}$$
* Jika $z < 0$, nilai berada di bawah rata-rata.
* Jika $z = 0$, nilai sama dengan rata-rata.
* Jika $z > 0$, nilai berada di atas rata-rata.

```python
# SEL CODE 2: Ukuran Variabilitas, Box Plot, dan z-Score
import numpy as np
import matplotlib.pyplot as plt

# Kumpulan Data Sampel GPAs
gpa_data = np.array([1.39, 1.76, 1.90, 2.12, 2.53, 2.71, 3.00, 3.33, 3.71, 4.00])

# 1. Perhitungan Variabilitas
range_val = np.ptp(gpa_data) # Peak-to-peak (max - min)
var_sample = np.var(gpa_data, ddof=1) # ddof=1 untuk sampel variance
std_sample = np.std(gpa_data, ddof=1) # ddof=1 untuk sampel std dev

print("--- Ukuran Penyebaran Data ---")
print(f"Range              : {range_val:.2f}")
print(f"Varians Sampel (s²): {var_sample:.4f}")
print(f"Standar Deviasi (s): {std_sample:.4f}")

# 2. Five-Number Summary & IQR
q1 = np.percentile(gpa_data, 25)
q2 = np.percentile(gpa_data, 50) # Median
q3 = np.percentile(gpa_data, 75)
iqr = q3 - q1

print("\n--- Five-Number Summary & IQR ---")
print(f"Min: {np.min(gpa_data)}, Q1: {q1}, Q2 (Median): {q2}, Q3: {q3}, Max: {np.max(gpa_data)}")
print(f"IQR: {iqr:.2f}")

# 3. Perhitungan z-Score untuk setiap data
z_scores = (gpa_data - np.mean(gpa_data)) / std_sample
print("\n--- z-Scores ---")
for x, z in zip(gpa_data, z_scores):
    print(f"Data: {x:.2f} -> z-score: {z:.2f}")

# 4. Visualisasi Box Plot
plt.figure(figsize=(8, 4))
plt.boxplot(gpa_data, vert=False, patch_artist=True, boxprops=dict(facecolor='lightgreen'))
plt.title('Box Plot Data GPA Mahasiswa', fontsize=12)
plt.xlabel('GPA', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

## 2.5. Aturan Empiris dan Teorema Chebyshev (Empirical Rule & Chebyshev's Theorem)

Kedua aturan ini memberikan prediksi proporsi data yang berada dalam rentang $k$ standar deviasi dari rata-rata ($\mu \pm k\sigma$ atau $\bar{x} \pm ks$).

### 1. Aturan Empiris (The Empirical Rule)
Berlaku **KHUSUS** untuk data yang memiliki distribusi simetris/berbentuk lonceng (*bell-shaped / normal distribution*):

* **$\approx 68\%$** data terletak dalam $1$ standar deviasi dari mean: $[\mu - \sigma, \mu + \sigma]$
* **$\approx 95\%$** data terletak dalam $2$ standar deviasi dari mean: $[\mu - 2\sigma, \mu + 2\sigma]$
* **$\approx 99.7\%$** data terletak dalam $3$ standar deviasi dari mean: $[\mu - 3\sigma, \mu + 3\sigma]$

---

### 2. Teorema Chebyshev (Chebyshev's Theorem)
Berlaku untuk **SEMUA jenis distribusi data** (apakah simetris maupun miring/skewed). Untuk setiap bilangan $k > 1$:

Proporsi data yang berada dalam $k$ standar deviasi dari mean ($\mu \pm k\sigma$) adalah **sekurang-kurangnya**:
$$1 - \frac{1}{k^2}$$

* Untuk $k = 2$: Minimal $1 - \frac{1}{2^2} = 1 - \frac{1}{4} = 75\%$ ($3/4$) data berada di interval $[\mu - 2\sigma, \mu + 2\sigma]$.
* Untuk $k = 3$: Minimal $1 - \frac{1}{3^2} = 1 - \frac{1}{9} \approx 88.89\%$ ($8/9$) data berada di interval $[\mu - 3\sigma, \mu + 3\sigma]$.

---

## 2.6. Contoh Soal & Pembahasan

### Contoh Soal 1: Penerapan Aturan Empiris
Diketahui skor tes IQ berdistribusi berbentuk lonceng dengan rata-rata $\mu = 100$ dan standar deviasi $\sigma = 10$.
1. Berapa persentase populasi yang memiliki skor IQ antara $80$ dan $120$?
2. Berapa persentase populasi yang memiliki skor IQ lebih besar dari $110$?

**Penyelesaian Step-by-Step**:
1. Batas $80$ dan $120$:
   * $80 = 100 - 2(10) = \mu - 2\sigma$
   * $120 = 100 + 2(10) = \mu + 2\sigma$
   * Berdasarkan **Aturan Empiris**, proporsi data dalam $[\mu - 2\sigma, \mu + 2\sigma]$ adalah sekitar **$95\%$**.

2. Skor IQ lebih besar dari $110$:
   * Batas $110 = 100 + 1(10) = \mu + 1\sigma$.
   * Berdasarkan Aturan Empiris, $68\%$ data berada antara $90$ ($\mu-\sigma$) dan $110$ ($\mu+\sigma$).
   * Sisanya adalah $100\% - 68\% = 32\%$ berada di luar rentang tersebut.
   * Karena kurva simetris, setengah dari $32\%$ berada di atas $110$, yaitu $\frac{32\%}{2} =$ **$16\%$**.

---

### Contoh Soal 2: Penerapan Teorema Chebyshev
Sebuah sampel berukuran $n = 50$ memiliki rata-rata $\bar{x} = 28$ dan standar deviasi $s = 3$. Tanpa mengetahui bentuk distribusi data, berapa jumlah minimum observasi yang berada dalam interval $(22, 34)$?

**Penyelesaian Step-by-Step**:
1. Hitung nilai $k$ dari interval $(22, 34)$:
   * $\bar{x} - k \cdot s = 22 \implies 28 - k(3) = 22 \implies 3k = 6 \implies k = 2$
   * $\bar{x} + k \cdot s = 34 \implies 28 + k(3) = 34 \implies 3k = 6 \implies k = 2$
2. Hitung proporsi minimum menggunakan Teorema Chebyshev:
   $$\text{Proporsi Minimum} = 1 - \frac{1}{k^2} = 1 - \frac{1}{2^2} = 0.75 \quad (75\%)$$
3. Hitung jumlah minimum observasi dari $n = 50$:
   $$\text{Jumlah Observasi} = 75\% \times 50 = 37.5$$
   Karena jumlah observasi harus berupa bilangan bulat, kita bulatkan ke atas menjadi sekurang-kurangnya **$38$ observasi**.

```python
# SEL CODE 3: Simulasi Aturan Empiris & Teorema Chebyshev
import numpy as np
import matplotlib.pyplot as plt

# Set seed untuk hasil yang konsisten
np.random.seed(42)

# Generate data distribusi normal (Bell-shaped)
mu, sigma = 100, 10
iq_scores = np.random.normal(mu, sigma, 10000)

# Menghitung persentase nyata di rentang mu - 2*sigma s.d. mu + 2*sigma
within_2sd = np.sum((iq_scores >= mu - 2*sigma) & (iq_scores <= mu + 2*sigma)) / len(iq_scores) * 100

print("--- Visualisasi & Pembuktian Empirical Rule ---")
print(f"Persentase aktual dalam [mu - 2*sigma, mu + 2*sigma]: {within_2sd:.2f}% (Teori: ~95%)")

# Plot Kurva Distribusi Normal dan Area Empirical Rule
plt.figure(figsize=(10, 5))
count, bins, ignored = plt.hist(iq_scores, 30, density=True, alpha=0.5, color='gray')
plt.plot(bins, 1/(sigma * np.sqrt(2 * np.pi)) * np.exp( - (bins - mu)**2 / (2 * sigma**2) ), linewidth=2, color='r')

# Arsir area 2 SD (80 s.d. 120)
plt.axvline(mu - 2*sigma, color='blue', linestyle='--', label='μ - 2σ (80)')
plt.axvline(mu + 2*sigma, color='blue', linestyle='--', label='μ + 2σ (120)')
plt.fill_between(bins, 0, 1/(sigma * np.sqrt(2 * np.pi)) * np.exp( - (bins - mu)**2 / (2 * sigma**2) ), 
                 where=(bins >= 80) & (bins <= 120), color='blue', alpha=0.2, label='Area 95% (Empirical Rule)')

plt.title('Distribusi Skor IQ (Aturan Empiris)', fontsize=14)
plt.xlabel('Skor IQ')
plt.ylabel('Densitas')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
