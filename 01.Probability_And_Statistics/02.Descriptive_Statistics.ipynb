## 1. Tiga Penyajian Data Populer
*   **Diagram Batang-Daun (Stem and Leaf Diagram)**: Menyajikan data dengan memisahkan nilai menjadi "batang" (digit puluhan) dan "daun" (digit satuan). Berguna untuk data berukuran kecil hingga sedang.
*   **Histogram Frekuensi**: Grafik batang yang menunjukkan frekuensi setiap kelas. Sumbu horizontal adalah kelas, sumbu vertikal adalah frekuensi.
*   **Histogram Frekuensi Relatif**: Sama seperti histogram frekuensi, tetapi sumbu vertikal menunjukkan frekuensi relatif (proporsi) dari setiap kelas. Total area semua batang sama dengan 1 atau 100%.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

# --- 1. TIGA PENYAJIAN DATA POPULER ---
print("=== 1. TIGA PENYAJIAN DATA POPULER ===")

# Contoh data nilai ujian 30 siswa
data_nilai = [86, 80, 25, 77, 73, 76, 100, 90, 69, 93, 90, 83, 70, 73, 73, 70, 90, 83, 71, 95, 40, 58, 68, 69, 100, 78, 87, 97, 92, 74]

# Stem and Leaf Diagram (sederhana)
print("Stem and Leaf Diagram:")
stems = {}
for nilai in data_nilai:
    stem = nilai // 10
    leaf = nilai % 10
    if stem not in stems:
        stems[stem] = []
    stems[stem].append(leaf)
for stem in sorted(stems.keys()):
    leaves = ' '.join(str(l) for l in sorted(stems[stem]))
    print(f"{stem:2d} | {leaves}")

# Frequency Histogram
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.hist(data_nilai, bins=range(0, 110, 10), edgecolor='black')
plt.title('Histogram Frekuensi')
plt.xlabel('Nilai')
plt.ylabel('Frekuensi')

# Relative Frequency Histogram
plt.subplot(1,2,2)
plt.hist(data_nilai, bins=range(0, 110, 10), edgecolor='black', density=True)
plt.title('Histogram Frekuensi Relatif')
plt.xlabel('Nilai')
plt.ylabel('Frekuensi Relatif')
plt.tight_layout()
plt.show()
print("\n")
```

## 2. Ukuran Lokasi Pusat (Tiga Jenis Rata-rata)
*   **Mean (Rata-rata)**: Jumlah semua nilai dibagi banyaknya nilai.
    *   Rata-rata sampel: $$\bar{x} = \frac{\sum x}{n}$$
    *   Rata-rata populasi: $$\mu = \frac{\sum x}{N}$$
*   **Median**: Nilai tengah setelah data diurutkan. Jika jumlah data genap, median adalah rata-rata dua nilai tengah.
    *   Notasi: $\tilde{x}$ untuk sampel, $\tilde{\mu}$ untuk populasi.
*   **Modus**: Nilai yang paling sering muncul. Data bisa memiliki satu modus, beberapa modus, atau tidak ada modus.
*   **Hubungan Mean, Median, Modus**:
    *   Distribusi simetris: $\bar{x} \approx \text{median} \approx \text{modus}$
    *   Distribusi miring ke kanan: $\bar{x} > \text{median} > \text{modus}$
    *   Distribusi miring ke kiri: $\bar{x} < \text{median} < \text{modus}$

```python
# --- 2. UKURAN LOKASI PUSAT ---
print("=== 2. UKURAN LOKASI PUSAT ===")

# Data contoh
data = [2, -1, 0, 2, 5, 7, 8, 3, 4, 6]

# Mean
mean_data = np.mean(data)
print(f"Mean (rata-rata) = {mean_data:.2f}")

# Median
median_data = np.median(data)
print(f"Median = {median_data:.2f}")

# Modus
mode_result = stats.mode(data)
print(f"Modus = {mode_result.mode} (muncul {mode_result.count} kali)")

# Contoh dengan data frekuensi
print("\nContoh data frekuensi:")
x = np.array([0, 1, 2, 3, 4])
f = np.array([3, 6, 6, 3, 1])
n = np.sum(f)
mean_freq = np.sum(x * f) / n
print(f"Mean dari tabel frekuensi = {mean_freq:.4f}")
print("\n")
```

## 3. Ukuran Variabilitas
*   **Range (Jangkauan)**: Selisih nilai maksimum dan minimum.
    $$R = x_{\text{max}} - x_{\text{min}}$$
*   **Variansi Sampel**:
    $$s^2 = \frac{\sum (x - \bar{x})^2}{n - 1}$$
    atau rumus singkat:
    $$s^2 = \frac{\sum x^2 - \frac{1}{n}(\sum x)^2}{n - 1}$$
*   **Standar Deviasi Sampel**:
    $$s = \sqrt{s^2}$$
*   **Variansi Populasi**:
    $$\sigma^2 = \frac{\sum (x - \mu)^2}{N}$$
*   **Standar Deviasi Populasi**:
    $$\sigma = \sqrt{\sigma^2}$$

```python
# --- 3. UKURAN VARIABILITAS ---
print("=== 3. UKURAN VARIABILITAS ===")

# Data contoh: Data Set II dari buku
data_var = np.array([46, 37, 40, 33, 42, 36, 40, 47, 34, 45])

# Range
range_data = np.max(data_var) - np.min(data_var)
print(f"Range = {range_data:.2f}")

# Variansi Sampel
var_sampel = np.var(data_var, ddof=1)
print(f"Variansi Sampel (s^2) = {var_sampel:.2f}")

# Standar Deviasi Sampel
std_sampel = np.std(data_var, ddof=1)
print(f"Standar Deviasi Sampel (s) = {std_sampel:.2f}")

# Variansi Populasi (asumsi data ini populasi)
var_pop = np.var(data_var, ddof=0)
std_pop = np.std(data_var, ddof=0)
print(f"Variansi Populasi (sigma^2) = {var_pop:.2f}")
print(f"Standar Deviasi Populasi (sigma) = {std_pop:.2f}")
print("\n")
```

## 4. Posisi Relatif Data
*   **Persentil**: Nilai $x$ adalah persentil ke-$P$ jika $P\%$ data kurang dari atau sama dengan $x$.
*   **Kuartil**: Membagi data menjadi empat bagian.
    *   $Q_1$ = kuartil pertama (persentil ke-25)
    *   $Q_2$ = median (persentil ke-50)
    *   $Q_3$ = kuartil ketiga (persentil ke-75)
*   **Jangkauan Antar Kuartil (IQR)**:
    $$IQR = Q_3 - Q_1$$
*   **Lima Serangkai (Five-Number Summary)**:
    $$\{x_{\text{min}}, Q_1, Q_2, Q_3, x_{\text{max}}\}$$
*   **Box Plot**: Diagram yang dibuat dari lima serangkai.
*   **Z-Score**: Mengukur berapa banyak standar deviasi suatu nilai dari mean.
    $$z = \frac{x - \mu}{\sigma} \quad \text{(populasi)} \quad \text{atau} \quad z = \frac{x - \bar{x}}{s} \quad \text{(sampel)}$$

```python
# --- 4. POSISI RELATIF DATA ---
print("=== 4. POSISI RELATIF DATA ===")

# Data contoh: 10 nilai GPA
gpa = np.array([1.90, 3.00, 2.53, 3.71, 2.12, 1.76, 2.71, 1.39, 4.00, 3.33])

# Persentil ke-80
p80 = np.percentile(gpa, 80)
print(f"Persentil ke-80 = {p80:.2f}")

# Kuartil
Q1 = np.percentile(gpa, 25)
Q2 = np.percentile(gpa, 50)
Q3 = np.percentile(gpa, 75)
print(f"Q1 = {Q1:.2f}, Q2 = {Q2:.2f}, Q3 = {Q3:.2f}")

# IQR
IQR = Q3 - Q1
print(f"IQR = {IQR:.2f}")

# Lima Serangkai
xmin = np.min(gpa)
xmax = np.max(gpa)
print(f"Lima Serangkai: {{{xmin:.2f}, {Q1:.2f}, {Q2:.2f}, {Q3:.2f}, {xmax:.2f}}}")

# Z-score (sampel)
mean_gpa = np.mean(gpa)
std_gpa = np.std(gpa, ddof=1)
z_scores = (gpa - mean_gpa) / std_gpa
print(f"Z-scores: {np.round(z_scores, 2)}")

# Box Plot
plt.figure()
plt.boxplot(gpa, vert=False)
plt.title('Box Plot GPA')
plt.xlabel('GPA')
plt.show()
print("\n")
```

## 5. Aturan Empiris dan Teorema Chebyshev
*   **Aturan Empiris**: Berlaku untuk distribusi berbentuk lonceng (normal).
    *   Sekitar 68% data berada dalam $\mu \pm \sigma$
    *   Sekitar 95% data berada dalam $\mu \pm 2\sigma$
    *   Sekitar 99.7% data berada dalam $\mu \pm 3\sigma$
*   **Teorema Chebyshev**: Berlaku untuk semua distribusi.
    *   Setidaknya $1 - \frac{1}{k^2}$ data berada dalam $\mu \pm k\sigma$ (untuk $k > 1$).
    *   Untuk $k=2$: setidaknya 75% data.
    *   Untuk $k=3$: setidaknya 88.9% data.

```python
# --- 5. ATURAN EMPIRIS DAN TEOREMA CHEBYSHEV ---
print("=== 5. ATURAN EMPIRIS DAN TEOREMA CHEBYSHEV ===")

# Contoh: Tinggi badan 100 pria dewasa, mean = 69.92, std = 1.70
mean_tinggi = 69.92
std_tinggi = 1.70

# Aturan Empiris
print("Aturan Empiris (untuk distribusi normal):")
print(f" 68% data dalam [{mean_tinggi - std_tinggi:.2f}, {mean_tinggi + std_tinggi:.2f}]")
print(f" 95% data dalam [{mean_tinggi - 2*std_tinggi:.2f}, {mean_tinggi + 2*std_tinggi:.2f}]")
print(f" 99.7% data dalam [{mean_tinggi - 3*std_tinggi:.2f}, {mean_tinggi + 3*std_tinggi:.2f}]")

# Teorema Chebyshev
print("\nTeorema Chebyshev (untuk semua distribusi):")
for k in [2, 3]:
    proporsi = 1 - 1/(k**2)
    lower = mean_tinggi - k*std_tinggi
    upper = mean_tinggi + k*std_tinggi
    print(f" Setidaknya {proporsi*100:.2f}% data dalam [{lower:.2f}, {upper:.2f}]")
print("\n")
```
