# **BAB 2 : DESCRIPTIVE STATISTICS**

Statistika deskriptif adalah cabang ilmu statistika yang berfokus pada pengorganisasian, penyajian, dan peringkasan data agar informasi penting di dalamnya dapat dipahami dengan mudah sebelum dilakukan analisis inferensial.



## 2.1. Visualisasi Data (Three Popular Data Displays)

Terdapat tiga bentuk visualisasi grafik yang sering digunakan untuk menggambarkan distribusi data kuantitatif:

1. **Diagram Batang dan Daun (*Stem and Leaf Diagram*)**
   * Memisahkan setiap nilai data menjadi dua bagian: **stem** (digit angka utama/puluhan) dan **leaf** (digit terakhir/satuan).
   * **Keunggulan**: Keutuhan seluruh data asli masih tetap terjaga dan dapat direkonstruksi kembali.

2. **Histogram Frekuensi (*Frequency Histogram*)**
   * Mengelompokkan data ke dalam beberapa kelas interval pada sumbu-$x$ dan menampilkan jumlah frekuensi pengamatan $f$ pada sumbu-$y$ dalam bentuk batang vertikal.

3. **Histogram Frekuensi Relatif (*Relative Frequency Histogram*)**
   * Mirip dengan histogram frekuensi, namun sumbu-$y$ menyatakan proporsi atau frekuensi relatif ($\frac{f}{n}$) dari setiap kelas.
   * Total luas daerah di bawah seluruh batang histogram relatif sama dengan $1$ (atau $100\%$). Jika ukuran sampel $n \to \infty$, bentuk histogram relatif akan mendekati kurva halus (*density curve*).



## 2.1. Contoh Soal & Pembahasan

**Soal:**
Diberikan sampel nilai ujian dari 10 mahasiswa: $75, 80, 80, 85, 85, 85, 90, 90, 95, 100$. Buatlah distribusi frekuensi relatif untuk data tersebut!

**Pembahasan:**
1. Hitung total sampel: $n = 10$.
2. Frekuensi relatif dihitung dengan rumus:
   $$\text{Frekuensi Relatif} = \frac{f}{n}$$
3. Tabel Frekuensi Relatif:
   * Nilai $75$: $f = 1 \implies \text{Frekuensi Relatif} = \frac{1}{10} = 0.10$
   * Nilai $80$: $f = 2 \implies \text{Frekuensi Relatif} = \frac{2}{10} = 0.20$
   * Nilai $85$: $f = 3 \implies \text{Frekuensi Relatif} = \frac{3}{10} = 0.30$
   * Nilai $90$: $f = 2 \implies \text{Frekuensi Relatif} = \frac{2}{10} = 0.20$
   * Nilai $95$: $f = 1 \implies \text{Frekuensi Relatif} = \frac{1}{10} = 0.10$
   * Nilai $100$: $f = 1 \implies \text{Frekuensi Relatif} = \frac{1}{10} = 0.10$
4. Total seluruh frekuensi relatif:
   $$0.10 + 0.20 + 0.30 + 0.20 + 0.10 + 0.10 = 1.00$$

```python
import matplotlib.pyplot as plt
import numpy as np

# Data nilai ujian
data_nilai = [75, 80, 80, 85, 85, 85, 90, 90, 95, 100]

# Membuat Histogram Frekuensi Relatif
plt.figure(figsize=(8, 5))
counts, bins, patches = plt.hist(data_nilai, bins=np.arange(70, 110, 5), 
                                 density=False, weights=np.ones(len(data_nilai)) / len(data_nilai),
                                 edgecolor='black', color='skyblue')

plt.title('Histogram Frekuensi Relatif Nilai Ujian')
plt.xlabel('Nilai Ujian')
plt.ylabel('Frekuensi Relatif')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

## 2.2. Ukuran Pemusatan Data (Measures of Central Location)

Ukuran pemusatan data bertujuan untuk menjawab pertanyaan: *"Di manakah titik pusat dari kumpulan data tersebut?"*

### 1. Rataan (*Mean*)
Rataan adalah jumlah seluruh nilai pengamatan dibagi dengan banyaknya pengamatan.

* **Sampel Mean ($\bar{x}$):**
  $$\bar{x} = \frac{\sum x}{n}$$
  Di mana $n$ adalah ukuran sampel.

* **Populational Mean ($\mu$):**
  $$\mu = \frac{\sum x}{N}$$
  Di mana $N$ adalah ukuran populasi.

### 2. Median ($\tilde{x}$)
Median adalah nilai tengah dari data yang telah diurutkan dari nilai terkecil ke terbesar.
* **Jumlah data ganjil ($n$ ganjil):** Nilai tepat di posisi tengah $\left(\frac{n+1}{2}\right)$.
* **Jumlah data genap ($n$ genap):** Rata-rata dari dua nilai tengah pada posisi $\left(\frac{n}{2}\right)$ dan $\left(\frac{n}{2} + 1\right)$.

*Catatan Kesentengan (Skewness):*
* **Simetris:** $\bar{x} = \tilde{x}$
* **Miring ke Kanan (*Skewed Right*):** $\bar{x} > \tilde{x}$ (ditarik oleh nilai ekstrem tinggi / *outlier* positif).
* **Miring ke Kiri (*Skewed Left*):** $\bar{x} < \tilde{x}$ (ditarik oleh nilai ekstrem rendah / *outlier* negatif).

### 3. Modus (*Mode*)
Modus adalah nilai yang paling sering muncul (memiliki frekuensi tertinggi) dalam sekumpulan data. Kumpulan data bisa bersifat unimodal, bimodal, multimodal, atau tidak memiliki modus sama sekali.


## 2.2. Contoh Soal & Pembahasan

**Soal:**
Hitunglah Mean, Median, dan Modus dari sampel data pendapatan bulanan (dalam juta rupiah) berikut: 
$$[2.4, 2.8, 2.5, 3.1, 2.8, 19.5, 2.1]$$

**Pembahasan:**
1. **Mean ($\bar{x}$):**
   $$\sum x = 2.4 + 2.8 + 2.5 + 3.1 + 2.8 + 19.5 + 2.1 = 35.2$$
   $$n = 7$$
   $$\bar{x} = \frac{35.2}{7} \approx 5.03$$

2. **Median ($\tilde{x}$):**
   Urutkan data dari terkecil ke terbesar:
   $$2.1, 2.4, 2.5, \mathbf{2.8}, 2.8, 3.1, 19.5$$
   Karena $n = 7$ (ganjil), median adalah data ke-$\frac{7+1}{2} = 4$, yaitu $2.8$.

3. **Modus:**
   Nilai $2.8$ muncul sebanyak 2 kali, sedangkan angka lainnya hanya 1 kali.
   $$\text{Modus} = 2.8$$

*Kesimpulan:* Nilai $19.5$ merupakan *outlier*, sehingga membuat Mean ($5.03$) terdistorsi ke kanan, sementara Median ($2.8$) lebih representatif menggambarkan pusat data.

```python
import numpy as np
from scipy import stats

data_pendapatan = [2.4, 2.8, 2.5, 3.1, 2.8, 19.5, 2.1]

# Perhitungan Mean, Median, Modus
mean_val = np.mean(data_pendapatan)
median_val = np.median(data_pendapatan)
mode_val = stats.mode(data_pendapatan, keepdims=True).mode[0]

print(f"Mean   : {mean_val:.2f}")
print(f"Median : {median_val:.2f}")
print(f"Modus  : {mode_val:.2f}")
```

## 2.3. Ukuran Penyebaran Data (Measures of Variability)

Ukuran penyebaran data digunakan untuk mengukur seberapa jauh data tersebar atau menyimpang dari titik pusatnya.

### 1. Rentang (*Range*)
Selisih antara nilai terbesar ($x_{\max}$) dan nilai terkecil ($x_{\min}$):
$$R = x_{\max} - x_{\min}$$

### 2. Varians Sampel ($s^2$) dan Deviasi Standar Sampel ($s$)
Mengukur rata-rata kuadrat jarak dari setiap data ke titik rata-rata ($\bar{x}$).

* **Varians Sampel ($s^2$):**
  $$s^2 = \frac{\sum (x - \bar{x})^2}{n - 1}$$
  *Atau menggunakan rumus pintas (*shortcut formula*):*
  $$s^2 = \frac{\sum x^2 - \frac{1}{n}\left(\sum x\right)^2}{n - 1}$$

* **Deviasi Standar Sampel ($s$):**
  $$s = \sqrt{s^2}$$

### 3. Varians Populasi ($\sigma^2$) dan Deviasi Standar Populasi ($\sigma$)
* **Varians Populasi:**
  $$\sigma^2 = \frac{\sum (x - \mu)^2}{N}$$
* **Deviasi Standar Populasi:**
  $$\sigma = \sqrt{\sigma^2}$$


## 2.3. Contoh Soal & Pembahasan

**Soal:**
Diberikan sampel data: $x = [2, 4, 6, 8]$. Hitunglah varians sampel ($s^2$) dan deviasi standar sampel ($s$)!

**Pembahasan:**
1. Hitung $\bar{x}$:
   $$\bar{x} = \frac{2 + 4 + 6 + 8}{4} = \frac{20}{4} = 5$$

2. Hitung jumlah kuadrat deviasi $\sum (x - \bar{x})^2$:
   * $(2 - 5)^2 = (-3)^2 = 9$
   * $(4 - 5)^2 = (-1)^2 = 1$
   * $(6 - 5)^2 = (1)^2 = 1$
   * $(8 - 5)^2 = (3)^2 = 9$
   $$\sum (x - \bar{x})^2 = 9 + 1 + 1 + 9 = 20$$

3. Hitung $s^2$ dan $s$ (dengan derajat kebebasan $n-1 = 3$):
   $$s^2 = \frac{20}{4 - 1} = \frac{20}{3} \approx 6.67$$
   $$s = \sqrt{6.67} \approx 2.58$$

```python
import numpy as np

data_x = np.array([2, 4, 6, 8])

# ddof=1 digunakan untuk menghitung sampel varians & std deviasi (n - 1)
variance_sample = np.var(data_x, ddof=1)
std_sample = np.std(data_x, ddof=1)

print(f"Varians Sampel (s^2) : {variance_sample:.2f}")
print(f"Deviasi Standar (s)  : {std_sample:.2f}")
```

## 2.4. Posisi Relatif Data (Relative Position of Data)

Posisi relatif menentukan di mana letak suatu nilai observasi spesifik jika dibandingkan dengan seluruh anggota kumpulan data.

### 1. Persentil dan Kuartil
* **Persentil ke-$P$ ($P^{th}\text{ Percentile}$):** Nilai di mana $P\%$ dari data bernilai kurang dari atau sama dengan nilai tersebut.
* **Kuartil ($Q_1, Q_2, Q_3$):** Membagi data terurut menjadi 4 bagian yang sama besar ($25\%$ tiap bagian).
  * $Q_1$ (Kuartil Pertama) = Median dari kelompok data bawah (persentil ke-25).
  * $Q_2$ (Kuartil Kedua) = Median dari seluruh data (persentil ke-50).
  * $Q_3$ (Kuartil Ketiga) = Median dari kelompok data atas (persentil ke-75).
* **Interquartile Range ($IQR$):**
  $$IQR = Q_3 - Q_1$$

### 2. Ringkasan Lima Angka (*Five-Number Summary*) & Box Plot
Ringkasan lima angka terdiri dari:
$$\{x_{\min},\ Q_1,\ Q_2,\ Q_3,\ x_{\max}\}$$
Elemen-elemen ini digambarkan secara visual dalam bentuk **Box Plot (Box and Whisker Plot)**.

### 3. Z-Score (Skor Standar)
$z$-score menyatakan berapa banyak standar deviasi suatu nilai observasi $x$ terletak di atas atau di bawah rata-rata:
$$z = \frac{x - \mu}{\sigma} \quad \text{atau} \quad z = \frac{x - \bar{x}}{s}$$


## 2.4. Contoh Soal & Pembahasan

**Soal:**
Suatu populasi IPK mahasiswa memiliki rata-rata $\mu = 2.70$ dan deviasi standar $\sigma = 0.50$. Jika IPK Antonio adalah $x = 3.34$, berapakah nilai $z$-score Antonio dan apa artinya?

**Pembahasan:**
1. Masukkan nilai ke dalam rumus $z$-score:
   $$z = \frac{x - \mu}{\sigma} = \frac{3.34 - 2.70}{0.50} = \frac{0.64}{0.50} = 1.28$$
2. **Interpretasi:** Nilai IPK Antonio ($3.34$) berada $1.28$ deviasi standar di atas rata-rata populasi.

```python
import matplotlib.pyplot as plt
import numpy as np

# Contoh data IPK
gpa_data = [1.39, 1.76, 1.90, 2.12, 2.53, 2.71, 3.00, 3.33, 3.71, 3.88, 4.00]

# Membuat Box Plot
plt.figure(figsize=(6, 4))
plt.boxplot(gpa_data, vert=False, patch_artist=True, boxprops=dict(facecolor="lightgreen"))
plt.title('Box Plot Data IPK Mahasiswa')
plt.xlabel('IPK')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()

# Menghitung Five-Number Summary
q1 = np.percentile(gpa_data, 25)
q2 = np.median(gpa_data)
q3 = np.percentile(gpa_data, 75)
iqr = q3 - q1

print(f"Min: {min(gpa_data)}, Q1: {q1}, Median (Q2): {q2}, Q3: {q3}, Max: {max(gpa_data)}")
print(f"Interquartile Range (IQR): {iqr:.2f}")
```

## 2.5. Aturan Empiris dan Teorema Chebyshev (The Empirical Rule and Chebyshev's Theorem)

Aturan ini menjelaskan estimasi proporsi data yang berada di dalam rentang interval tertentu dari rata-rata ($\bar{x} \pm k\cdot s$).

### 1. Aturan Empiris (*Empirical Rule*)
*Hanya berlaku untuk data yang berdistribusi SIMETRIS dan BERBENTUK LONCENG (*Bell-shaped*).*

* **$\bar{x} \pm 1s$** ($\mu \pm 1\sigma$): Mencakup sekitar **$68\%$** dari seluruh data.
* **$\bar{x} \pm 2s$** ($\mu \pm 2\sigma$): Mencakup sekitar **$95\%$** dari seluruh data.
* **$\bar{x} \pm 3s$** ($\mu \pm 3\sigma$): Mencakup sekitar **$99.7\%$** dari seluruh data.

### 2. Teorema Chebyshev (*Chebyshev's Theorem*)
*Berlaku untuk SEMUA jenis distribusi data (termasuk yang miring/asimetris).*

Untuk nilai $k > 1$, proporsi minimal data yang berada di dalam interval $\bar{x} \pm k\cdot s$ adalah:
$$\text{Proporsi Minimal} = 1 - \frac{1}{k^2}$$

* **Untuk $k = 2$:** Setidaknya $1 - \frac{1}{2^2} = 1 - \frac{1}{4} = \mathbf{75\%}$ data berada di dalam rentang $(\bar{x} - 2s, \bar{x} + 2s)$.
* **Untuk $k = 3$:** Setidaknya $1 - \frac{1}{3^2} = 1 - \frac{1}{9} \approx \mathbf{88.89\%}$ data berada di dalam rentang $(\bar{x} - 3s, \bar{x} + 3s)$.


## 2.5. Contoh Soal & Pembahasan

**Soal:**
Suatu sampel dengan $n = 50$ memiliki rata-rata $\bar{x} = 28$ dan deviasi standar $s = 3$. Tanpa mengetahui bentuk distribusi data, berapa jumlah minimum observasi yang dijamin berada dalam interval $(22, 34)$?

**Pembahasan:**
1. Tentukan batas interval dalam bentuk $\bar{x} \pm k\cdot s$:
   $$22 = 28 - 2(3) \implies \bar{x} - 2s$$
   $$34 = 28 + 2(3) \implies \bar{x} + 2s$$
   Maka, nilai $k = 2$.

2. Gunakan **Teorema Chebyshev** (karena bentuk distribusi tidak diketahui):
   $$\text{Proporsi Minimal} = 1 - \frac{1}{2^2} = 0.75 \ (75\%)$$

3. Hitung jumlah observasi minimum:
   $$\text{Jumlah minimum} = 75\% \times 50 = 37.5$$
   Karena jumlah observasi harus berupa bilangan bulat, kita bulatkan ke atas.
   $$\text{Jumlah minimum observasi} = 38 \text{ data}$$

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Visualisasi Aturan Empiris (Distribusi Normal / Bell-shaped)
mu = 69.92
sigma = 1.70

x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
y = stats.norm.pdf(x, mu, sigma)

plt.figure(figsize=(9, 5))
plt.plot(x, y, color='black', linewidth=2)

# Shade 1 Std Dev (68%)
plt.fill_between(x, y, where=(x >= mu - sigma) & (x <= mu + sigma), color='skyblue', alpha=0.6, label='68% (mu ± 1 sigma)')
# Shade 2 Std Dev (95%)
plt.fill_between(x, y, where=(x >= mu - 2*sigma) & (x <= mu + 2*sigma), color='blue', alpha=0.2, label='95% (mu ± 2 sigma)')

plt.title('Visualisasi Aturan Empiris (Empirical Rule)')
plt.xlabel('Nilai')
plt.ylabel('Densitas Probabilitas')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
