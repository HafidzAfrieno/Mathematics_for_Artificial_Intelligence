## 1. Relasi dan Fungsi
- Relasi adalah hubungan antara anggota suatu himpunan domain $X$ dengan himpunan kodomain $Y$.
- Fungsi ($f: X \to Y$) adalah relasi khusus di mana setiap elemen $x \in X$ terhubung tepat dengan satu elemen $y \in Y$, yang ditulis sebagai $y = f(x)$.

```python
import numpy as np
import math
def fungsi_kuadrat(x):
    return x**2 - 4*x + 4

x_val = 5
y_val = fungsi_kuadrat(x_val)
print(f"Hasil pemetaan fungsi f({x_val}) = {y_val:.2f}\n")
```

## 2. Gradien dan Persamaan Garis Lurus
- Gradien ($m$) menyatakan kemiringan garis lurus yang melalui dua titik $A(x_1, y_1)$ dan $B(x_2, y_2)$:$$m = \frac{y_2 - y_1}{x_2 - x_1}$$
- Persamaan Garis Lurus (Bentuk Eksplisit):$$y - y_1 = m(x - x_1) \quad \text{atau} \quad y = mx + c$$
- Persamaan Linier Umum / Hyperplane (Bentuk Implisit):Dalam ruang berdimensi-$n$, persamaannya dinotasikan sebagai:$$w_1 x_1 + w_2 x_2 + \dots + w_n x_n + b = 0 \quad \iff \quad \mathbf{w}^T \mathbf{x} + b = 0$$

```python
def hitung_gradien_dan_persamaan(x1, y1, x2, y2):
    m = (y2 - y1) / (x2 - x1)
    c = y1 - m * x1
    return m, c

x1, y1 = 2, 3
x2, y2 = 4, 8
m, c = hitung_gradien_dan_persamaan(x1, y1, x2, y2)
print(f"Gradien garis AB (m): {m:.2f}")
print(f"Persamaan garis: y = {m:.2f}x + ({c:.2f})\n")
```

## 3. Grafik Sepotong-sepotong (Piecewise Function)
Fungsi sepotong-sepotong didefinisikan oleh beberapa sub-fungsi yang berlaku pada domain/interval berbeda:
$$f(x) = \begin{cases} f_1(x), & \text{jika } x < a \\ f_2(x), & \text{jika } a \le x \le b \\ f_3(x), & \text{jika } x > b \end{cases}$$

```python
def fungsi_piecewise(x):
    if x < 0:
        return -x
    elif 0 <= x <= 5:
        return x**2
    else:
        return 2*x + 15

print(f"Nilai piecewise f(-3) : {fungsi_piecewise(-3):.2f}")
print(f"Nilai piecewise f(3)  : {fungsi_piecewise(3):.2f}")
print(f"Nilai piecewise f(7)  : {fungsi_piecewise(7):.2f}\n")
```

## 4. Fungsi Keanggotaan Fuzzy (Fuzzy Membership Function)
Dalam logika Fuzzy, tingkat keanggotaan $\mu(x)$ berada pada rentang $[0, 1]$. Salah satu bentuk paling umum adalah Fuzzy Kurva Segitiga:
$$\mu(x) = \begin{cases} 0, & x \le a \text{ atau } x \ge c \\ \frac{x - a}{b - a}, & a < x \le b \\ \frac{c - x}{c - b}, & b < x < c \end{cases}$$

```python
def fuzzy_segitiga(x, a, b, c):
    if x <= a or x >= c:
        return 0.0
    elif a < x <= b:
        return (x - a) / (b - a)
    elif b < x < c:
        return (c - x) / (c - b)

a, b, c = 10, 25, 40
x_test = 20
mu_x = fuzzy_segitiga(x_test, a, b, c)
print(f"Derajat keanggotaan fuzzy mu({x_test}) pada [a={a}, b={b}, c={c}]: {mu_x:.3f}\n")
```

## 5. Hyperplane pada Support Vector Machine (SVM)
SVM mencari hyperplane pemisah optimal antara dua kelas $y_i \in \{-1, 1\}$ dengan memaksimalkan margin:
$$\mathbf{w}^T \mathbf{x} + b = 0$$
Aturan keputusan klasifikasinya adalah:
$$\hat{y} = \text{sign}(\mathbf{w}^T \mathbf{x} + b)$$

```python
def prediksi_svm(x, w, b):
    # f(x) = w1*x1 + w2*x2 + b
    skor = np.dot(w, x) + b
    kelas = 1 if skor >= 0 else -1
    return skor, kelas

# Bobot w1=0.5, w2=1.5, bias b=-4
w = np.array([0.5, 1.5])
b = -4.0
data_point = np.array([5, 8])

skor, kelas = prediksi_svm(data_point, w, b)
print(f"Titik data: {data_point}")
print(f"Nilai Hyperplane SVM: {skor:.2f}")
print(f"Prediksi Kelas SVM : C{1 if kelas == -1 else 2} (Nilai Label: {kelas})\n")
```

## 6. Fungsi Aktivasi pada Jaringan Saraf Tiruan (ANN)
Perceptron menerima masukan terbobot $z = \mathbf{w}^T \mathbf{x} + b$ dan mentransformasikannya menggunakan fungsi aktivasi linier atau non-linier.
- Fungsi Identitas (Linier): $f(z) = z$
- Fungsi ReLU (Rectified Linear Unit):$$f(z) = \max(0, z)$$

```python
def relu(z):
    return max(0.0, z)

# Input x, Bobot w, Bias b
inputs = np.array([2.0, 3.0])
weights = np.array([0.8, -0.5])
bias = 0.2

z = np.dot(inputs, weights) + bias
output_activation = relu(z)

print(f"Nilai Terbobot (z) : {z:.3f}")
print(f"Output ReLU        : {output_activation:.3f}\n")
```

### 3. Fungsi Sepotong-Sepotong

Suatu fungsi sepotong-sepotong didefinisikan sebagai berikut:

$$
f(x) = 
\begin{cases} 
2x + 1, & x < 2 \\ 
x^2 - 3, & 2 \le x \le 6 \\ 
5x - 10, & x > 6 
\end{cases}
$$

Hitunglah nilai dari $f(-1) + f(4) - f(7)$!


```python
def piecewise_func(x):
    if x < 2:
        return 2 * x + 1
    
    elif x >= 2 and x <= 6:
        return x ** 2 - 3
    
    elif x > 6:
        return 5 * x - 10

x_func = [-1,4,7]
result = []
for i in range(len(x_func)):
    result.append(piecewise_func(x_func[i]))

result_func = result[0] + result[1] - result[2]
print(f'nilai dari f(−1) + f(4) − f(7) adalah {result_func}')
```

### 4. Dalam sebuah sistem logika fuzzy, fungsi keanggotaan suhu "Hangat" berbentuk kurva segitiga dengan batas-batas parameter a=20^∘ "C"" b=30^∘ "C",", dan c=40^∘ "C"". Hitunglah derajat keanggotaa μ(x)"untuk suhu x=24^∘ "C""da x=35^∘ "C""!

```python
suhu_a = 20
suhu_b = 30
suhu_c = 40
x_fuzzy = [24,35]
result_fuzzy = []

def fuzz_logic(x):
    if x <= suhu_a or x >= suhu_c:
        return 0
    elif  x > suhu_a and x <= suhu_b:
        return (x - suhu_a) / (suhu_b - suhu_a)
    else:
        return (suhu_c - x) / (suhu_c - suhu_b)

for i in range(len(x_fuzzy)):
    output_fzzy = fuzz_logic(x_fuzzy[i])
    result_fuzzy.append(output_fzzy)
    print(f'Keangotaan dari {x_fuzzy[i]}C adalah {result_fuzzy[i]}')
```

### 7. Sebuah neuron pada Jaringan Saraf Tiruan memiliki masukan x=[3,4,1]dengan vektor bobot w=[0.5,−0.2,1.2]dan bias b=−0.5. Hitunglah nilai kombinasi linier z=w^T x+bdan tentukan output aktivasi jika fungsi aktivasi yang digunakan adalah fungsi ReLU!

```python
x_input = [3,4,1]
weight = [0.5,-0.2,1.2]
bias = -0.5

def relu_func(x):
    if x < 0:
        return 0
    else:
        return x
    
def count_neuron(x,w,b):
    count = 0
    for i in range(len(x)):
        count += x[i] * w[i]
    return relu_func(count + b)

ouput_neuron = count_neuron(x=x_input,w=weight,b=bias)
print(f'output aktivasi jika fungsi aktivasi Relu adalah {ouput_neuron}')
```
